## 3. الشرح المتدرج

### 🟢 أ) مساعد برمجي (Code Assistant) — نمط Cursor/GitHub Copilot


````python
class CodeAssistantAgent:
    """
    معمارية مرجعية لمساعد برمجي: فهم السياق (المشروع الكامل) + تنفيذ + تحقق
    """
    def __init__(self, llm, codebase_rag: SimpleRAGPipeline, code_executor):
        self.llm = llm
        self.codebase_rag = codebase_rag   # فهرسة الكود الموجود كمستندات RAG
        self.code_executor = code_executor

    def generate_code_with_context(self, instruction: str, current_file: str) -> dict:
        # 1. استرجاع أجزاء ذات صلة من بقية المشروع (دوال مشابهة، تعريفات أنواع)
        relevant_code = self.codebase_rag.retrieve(instruction, top_k=5)

        prompt = f"""أنت مساعد برمجي خبير. اتّبع نمط الكود الموجود في المشروع بدقة.

أجزاء ذات صلة من المشروع:
{[c['text'] for c in relevant_code]}

الملف الحالي:
{current_file}

المطلوب: {instruction}

اكتب الكود المطلوب فقط، متّسقًا مع أسلوب المشروع:"""
        generated_code = self.llm.generate(prompt)

        # 2. تحقق تلقائي: تشغيل الكود واختباره قبل تقديمه
        test_result = self.code_executor.run_with_tests(generated_code)
        if not test_result["passed"]:
            fix_prompt = f"""الكود التالي فشل في الاختبار:
{generated_code}
خطأ: {test_result['error']}
صحّح الكود:"""
            generated_code = self.llm.generate(fix_prompt)

        return {"code": generated_code, "verified": test_result["passed"]}
````

### 🟡 ب) روبوت دعم عملاء (Customer Support Bot) — نمط متكامل RAG + Agent + Escalation


````python
class CustomerSupportAgent:
    """
    معمارية شاملة: RAG على قاعدة المعرفة + أدوات للتفاعل مع أنظمة حقيقية
    (تتبع الطلبات، إلخ) + تصعيد ذكي لموظف بشري عند الحاجة
    """
    def __init__(self, llm, knowledge_base_rag, order_system_tool, escalation_threshold=0.6):
        self.llm = llm
        self.kb_rag = knowledge_base_rag
        self.order_tool = order_system_tool
        self.escalation_threshold = escalation_threshold

    def handle_query(self, user_query: str, user_id: str) -> dict:
        # 1. تقييم أولي: هل هذا سؤال معرفي عام، أم يتطلب بيانات خاصة بالمستخدم؟
        intent_prompt = f"""صنّف نوع الاستفسار التالي:
الاستفسار: {user_query}
الأنواع الممكنة: سؤال_عام، استعلام_طلب، شكوى، طلب_استرجاع
أعطِ التصنيف فقط."""
        intent = self.llm.generate(intent_prompt).strip()

        if intent == "استعلام_طلب":
            order_info = self.order_tool.get_order_status(user_id)
            response = self.llm.generate(f"معلومات الطلب: {order_info}\nصُغ ردًا ودودًا: ")
        else:
            retrieved = self.kb_rag.retrieve_with_reranking(user_query, final_k=3)
            response = self.llm.generate(
                self.kb_rag.build_augmented_prompt(user_query, retrieved)
            )

        # 2. تقييم الثقة وقرار التصعيد
        confidence_prompt = f"""السؤال: {user_query}\nالإجابة المُقترَحة: {response}
ما مدى ثقتك بأن هذه الإجابة صحيحة وكافية؟ أعطِ رقمًا من 0 إلى 1 فقط."""
        try:
            confidence = float(self.llm.generate(confidence_prompt).strip())
        except ValueError:
            confidence = 0.5

        if confidence < self.escalation_threshold or intent == "شكوى":
            return {
                "response": response, "escalated": True,
                "escalation_reason": "ثقة منخفضة" if confidence < self.escalation_threshold else "شكوى تتطلب تدخلًا بشريًا"
            }
        return {"response": response, "escalated": False, "confidence": confidence}
````

### 🔴 ج) طبقة التعمّق: نمط "Copilot" العام القابل للتعميم على أي مجال


````python
class GenericCopilotPattern:
    """
    النمط المعماري العام الذي تتشارك فيه كل "مساعدات Copilot" الناجحة:
    1. فهم السياق المحلي (الملف الحالي/المستند/الشاشة)
    2. فهم السياق العام (قاعدة المعرفة/المشروع الكامل عبر RAG)
    3. اقتراح فعل محدد وصغير (وليس تغييرًا جذريًا دفعة واحدة)
    4. تحقق تلقائي قبل العرض على المستخدم
    5. تغذية راجعة فورية (قبول/رفض المستخدم) تُحسِّن النظام لاحقًا
    """
    def __init__(self, llm, local_context_fn, global_rag, verifier_fn):
        self.llm = llm
        self.local_context_fn = local_context_fn      # يستخرج السياق المحلي الحالي
        self.global_rag = global_rag                    # معرفة عامة أوسع
        self.verifier_fn = verifier_fn                  # دالة تحقق مخصصة بالمجال
        self.feedback_log = []

    def suggest(self, trigger_context: dict) -> dict:
        local_ctx = self.local_context_fn(trigger_context)
        global_ctx = self.global_rag.retrieve(str(local_ctx), top_k=3)

        suggestion = self.llm.generate(f"""السياق المحلي: {local_ctx}
السياق العام ذو الصلة: {global_ctx}
اقترح الفعل التالي الأنسب والأكثر دقة:""")

        verified, verification_info = self.verifier_fn(suggestion)
        return {"suggestion": suggestion, "verified": verified, "details": verification_info}

    def log_feedback(self, suggestion_id: str, accepted: bool):
        self.feedback_log.append({"id": suggestion_id, "accepted": accepted})
        # في نظام ناضج: تُستخدم هذه البيانات لاحقًا لـ fine-tuning دوري (الفصل 23/30)
````

## 4. مثال محلول خطوة بخطوة

**الحالة:** روبوت دعم لعملاء شركة اتصالات عربية، يستقبل 1000 تذكرة يوميًا.

**توزيع النوايا:**

| النية | النسبة | المسار |
|---|---:|---|
| استفسار عن باقة | 45% | RAG على قاعدة المعرفة |
| حالة طلب/فاتورة | 30% | أداة CRM |
| شكوى | 15% | تصعيد بشري |
| سؤال عام | 10% | RAG + إجابة عامة |

**النتائج المقترحة:**

| المؤشر | القيمة |
|---|---:|
| نسبة الحل الآلي (Containment) | 65% |
| التصعيد البشري | 35% |
| رضا العملاء (CSAT) | 4.1/5 |
| زمن أول استجابة | 3 ثوانٍ |
| تكلفة التذكرة الآلية | 0.02 دولار |
| تكلفة التذكرة البشرية | 2.50 دولار |

**حساب التوفير اليومي:** 650 تذكرة آلية × 2.50 = 1625 دولارًا كانت ستُصرف بشريًا، مقابل 650 × 0.02 = 13 دولارًا تكلفة آلية. التوفير ≈ 1612 دولارًا يوميًا، أي نحو 48 ألف دولار شهريًا. لكن هذا الرقم لا يكتمل دون طرح تكلفة بناء النظام وتشغيله ومراجعة الجودة.

**الدرس:** المقاييس التقنية وحدها لا تكفي؛ يجب ربطها بمقاييس أعمال (Containment، CSAT، تكلفة التذكرة) لاتخاذ قرار.


## 5. نصائح وحيل

> 💡 **اضبط النطاق**: ابدأ بمجال ضيق (باقة واحدة، نوع تذكرة واحد) وتوسّع بعد إثبات الجودة. الروبوت العام يفشل في كل شيء.

> 💡 **اجعل الاقتباس إلزاميًا في المجالات الحساسة**: القانوني والطبي والمالي يحتاجان مصدرًا مرئيًا لكل ادعاء. «الثقة» بلا مصدر ليست كافية.

> 💡 **صمّم التصعيد كمسار نجاح لا فشل**: عندما يقول النظام «سأحوّلك لموظف»، فهذا أفضل من إجابة خاطئة. قِس معدل التصعيد وجودته.

> 💡 **اجمع التغذية الراجعة على مستوى الاقتراح**: قبول/رفض، تعديل، وتعليق. هذه البيانات أثمن من أي مجموعة تدريب عامة.

> 💡 **ابنِ مجموعة تقييم من مجال العميل نفسه**: أسئلة حقيقية، لهجات، أخطاء إملائية، ومصطلحات داخلية. التقييم العام لا يكشف فشل المجال.


## 6. أخطاء شائعة ومصائد

> ⚠️ **بناء «ChatGPT عام» للمجال**: واجهة محادثة بلا معرفة متخصصة وأدوات لا تحل مشكلة العميل. ابدأ من المهمة لا من التقنية.

> ⚠️ **لا مسار تصعيد**: النظام يحاول الإجابة في كل شيء، فيخترع سياسات أو أرقامًا. التصعيد الواضح يحمي السمعة.

> ⚠️ **هلوسة السياسات والأسعار**: معلومات متغيرة يجب أن تأتي من مصدر حي (API/قاعدة بيانات)، لا من ذاكرة النموذج.

> ⚠️ **تجاهل اللهجات والأخطاء الإملائية**: العربية ليست لغة واحدة في الاستخدام. اختبر على لهجات العميل الفعلية، لا على الفصحى فقط.

> ⚠️ **تسريب بيانات بين العملاء**: في نظام متعدد المستأجرين، فلترة الصلاحيات يجب أن تكون في طبقة الاسترجاع والأدوات، لا في الـ prompt فقط.


## 7. جدول مقارنة

| نوع التطبيق | المكوّنات الأساسية | التحدي الخاص |
|---|---|---|
| **مساعد برمجي** | RAG على الكود + تنفيذ واختبار تلقائي | فهم بنية مشروع كبير ضمن حدود السياق |
| **دعم عملاء** | RAG + أدوات أنظمة خارجية + تصعيد ذكي | التوازن بين الأتمتة والحاجة لتدخل بشري |
| **تحليل مستندات قانونية/طبية** | RAG دقيق جدًا + Citations إلزامية | دقة قصوى، لا مجال للهلوسة |
| **وكيل بحث (Research Agent)** | Multi-step RAG + تجميع وتركيب من مصادر متعددة | الموثوقية عبر مصادر متضاربة |
| **مساعد كتابة إبداعية** | ذاكرة أسلوب المستخدم + تكرار وتحسين تفاعلي | الحفاظ على صوت/أسلوب متّسق |


## 8. ملخص أفضل الممارسات

- صمّم دائمًا مسار **تصعيد واضح** لحالات عدم اليقين — لا تدع النظام "يُخمِّن" في سياقات حرجة.
- اجمع تغذية راجعة من المستخدم الحقيقي (قبول/رفض الاقتراحات) كمصدر بيانات لا يُقدَّر بثمن لتحسينات مستقبلية.
- ابدأ بأبسط نسخة ممكنة من كل تطبيق (RAG بسيط، وكيل واحد)، وأضف التعقيد (multi-agent، ذاكرة متقدمة) تدريجيًا فقط بناءً على فجوات أداء مُقاسة فعليًا.
- اختبر كل تطبيق بنفس صرامة الفصل التاسع (تقييم، سلامة) قبل أي نشر حقيقي.

## 9. أسئلة مقابلات

**سؤال 1: كيف تكيّف LLM لمجال متخصص؟**

> **إجابة نموذجية:** أبدأ بتحديد المهمة والمخاطر، ثم أبني طبقة معرفة (RAG) على وثائق المجال مع صلاحيات واقتباسات. أضيف أدوات للأنظمة الحية عند الحاجة، وأضبط الـ prompt بمصطلحات المجال وأمثلة حقيقية. إذا كانت الجودة غير كافية، أنتقل إلى Fine-tuning خفيف (LoRA) على بيانات المجال. التقييم بمجموعة ذهبية من خبراء المجال هو الفيصل، وليس حجم النموذج.

**سؤال 2: كيف تقيّم تطبيقًا متخصصًا؟**

> **إجابة نموذجية:** ثلاثة مستويات: تقني (دقة الاسترجاع، Faithfulness، زمن الاستجابة)، مجال (صحة المصطلحات، الالتزام بالسياسات، اكتمال الإجابة)، وأعمال (Containment، CSAT، تكلفة التذكرة، معدل التصعيد). أضيف عينة مراجعة بشرية أسبوعية، وأتابع الحالات الحرجة. التقييم يجب أن يكون مستمرًا لأن المعرفة والسياسات تتغير.

**سؤال 3: متى تبني نموذجًا خاصًا بدل استخدام API؟**

> **إجابة نموذجية:** عندما تكون الخصوصية أو الامتثال أو التكلفة على المدى الطويل أو الحاجة إلى تحكم كامل في المعمارية هي العامل الحاسم، ويكون لديك فريق وتحتية قادران على التشغيل. إذا كانت المهمة عامة والحجم صغيرًا، فالـ API أسرع وأرخص. قرار «بناء vs شراء» يعتمد على حجم الاستخدام، حساسية البيانات، والتمايز التنافسي، لا على الاتجاه السائد.
