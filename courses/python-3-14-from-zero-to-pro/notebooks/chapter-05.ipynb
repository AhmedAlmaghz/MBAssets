## 5.2 العمليات الحسابية الأساسية

````python
a: int = 10
b: int = 3

print(a + b)   # 13   ← الجمع
print(a - b)   # 7    ← الطرح
print(a * b)   # 30   ← الضرب
print(a / b)   # 3.3333333333333335  ← القسمة (تُعيد دائمًا float!)
````

### ⚠️ تحذير مهم: القسمة `/` تُعيد دائمًا `float`

````python
result = 10 / 2
print(result)        # 5.0  ← float! ليس 5 كـ int
print(type(result))  # <class 'float'>
````

> 💡 **لماذا هذا مهم؟** في لغات كثيرة أخرى، قسمة عددين صحيحين تُعيد عددًا صحيحًا. لكن Python 3 غيّرت هذا السلوك عمدًا (مقارنة بـ Python 2 القديمة) لتجنّب فقدان الدقة العشرية بشكل غير متوقع.

## 5.3 العمليات الحسابية المتقدمة

### ➗ القسمة الصحيحة (Floor Division) `//`

````python
print(10 // 3)   # 3   ← ناتج القسمة بدون الباقي (مُقرَّب للأسفل)
print(10 / 3)    # 3.3333...  ← القسمة العادية للمقارنة
````

### 🔢 باقي القسمة (Modulo) `%`

````python
print(10 % 3)    # 1   ← الباقي بعد القسمة
print(20 % 5)    # 0   ← قابل للقسمة تمامًا
````

> 🎯 **استخدام عملي شديد الشيوع**: `%` هي الأداة الأساسية للتحقق من كون رقم زوجيًا أو فرديًا:

````python
number: int = 7
if number % 2 == 0:
    print("زوجي")
else:
    print("فردي")
````

### 🔺 الأُس (Exponentiation) `**`

````python
print(2 ** 3)    # 8    ← 2 أُس 3
print(10 ** 2)   # 100  ← 10 تربيع
print(4 ** 0.5)  # 2.0  ← الجذر التربيعي (أُس 0.5)
````

> ⚠️ **خطأ شائع عند القادمين من لغات أخرى**: في بعض اللغات (مثل JavaScript القديمة)، الأُس يُكتب كدالة `Math.pow()`. في Python، استخدم `**` مباشرة - لا تخلط مع `^` التي تعني شيئًا مختلفًا تمامًا في Python (XOR على مستوى البت - موضوع متقدم لن نحتاجه الآن).

### 📊 جدول شامل لكل العمليات الحسابية

| العامل | الاسم | مثال | النتيجة |

| `+` | جمع | `7 + 3` | `10` |
| `-` | طرح | `7 - 3` | `4` |
| `*` | ضرب | `7 * 3` | `21` |
| `/` | قسمة عادية | `7 / 3` | `2.333...` |

| `%` | باقي القسمة | `7 % 3` | `1` |
| `**` | أُس | `7 ** 3` | `343` |

## 5.4 أولوية العمليات (Operator Precedence)

### 📐 القاعدة المعروفة من الرياضيات (PEMDAS)

Python تتبع نفس القواعد الرياضية التقليدية التي تعلمتها في المدرسة:

```
1. الأقواس ()
2. الأُس **
3. الضرب * والقسمة / // % (من اليسار لليمين)
4. الجمع + والطرح - (من اليسار لليمين)
```

````python
result = 2 + 3 * 4
print(result)  # 14  ← الضرب أولاً (3*4=12)، ثم الجمع (2+12=14)

result = (2 + 3) * 4
print(result)  # 20  ← الأقواس أولاً (2+3=5)، ثم الضرب (5*4=20)
````

> ✅ **نصيحة احترافية**: حتى عندما تكون متأكدًا من أولوية العمليات، استخدم الأقواس `()` لجعل الكود **أوضح** لأي قارئ آخر (أو لك مستقبلاً). الوضوح دائمًا أهم من توفير حرفين من الكتابة:

```python
# يعمل لكن يحتاج تفكيرًا لفهمه
total = price * quantity - discount * quantity

# أوضح بكثير بنفس النتيجة
total = (price * quantity) - (discount * quantity)
```

🧾 **للقراءة فقط** — مثال مقتطف يعتمد على `discount`، `price`، `quantity` من مثال سابق.

## 5.5 فخ الدقة العشرية: المشكلة الأشهر في كل لغات البرمجة

### 🔬 التجربة الصادمة

````python
print(0.1 + 0.2)
# الناتج: 0.30000000000000004
# ليس 0.3 كما تتوقع!
````

هذا ليس خطأ في Python - **كل** لغات البرمجة تقريبًا (C++, Java, JavaScript) تعاني من نفس المشكلة بالضبط.

### 🤔 لماذا يحدث هذا؟

الكمبيوتر يُخزّن كل شيء في النظام **الثنائي** (0 و 1 فقط). بعض الكسور العشرية البسيطة في نظامنا العشري (مثل 0.1) لا يمكن تمثيلها **بدقة تامة** في النظام الثنائي - تمامًا كما أن الكسر 1/3 لا يمكن كتابته بدقة تامة كعدد عشري (0.3333... إلى ما لا نهاية).

> 💡 **تشبيه مبسّط**: تخيل أنك تحاول قياس طول بمسطرة مُقسّمة لأثلاث فقط (لا أرباع ولا أخماس). بعض الأطوال ستقيسها بدقة تامة، لكن أطوالاً أخرى ستحتاج تقريبًا. هذا بالضبط ما يحدث مع الكمبيوتر والنظام الثنائي مقابل بعض الكسور العشرية.

### ⚠️ تبعات عملية خطيرة لهذه المشكلة

````python
price: float = 19.99
quantity: int = 3
total: float = price * quantity

print(total)  # 59.96999999999999  ← ليس 59.97 بالضبط!
````

تخيل لو كان هذا نظام حسابات بنكية حقيقي - هذا الفرق الصغير جدًا قد يُسبب مشاكل محاسبية خطيرة متراكمة.

### ✅ الحل الأول: التقريب عند العرض باستخدام `round()`

````python
total: float = 0.1 + 0.2
print(round(total, 2))  # 0.3  ← تقريب لخانتين عشريتين
````

````python
price: float = 19.99
quantity: int = 3
total: float = round(price * quantity, 2)
print(total)  # 59.97
````

> ⚠️ **تحذير مهم**: `round()` يحل مشكلة **العرض**، لكن لا يحل مشكلة **الدقة الداخلية** أثناء الحسابات المتتالية. في عمليات حسابية طويلة ومتكررة (كأنظمة مالية حقيقية)، لا تعتمد فقط على `round()`.

### ✅ الحل الثاني (الاحترافي): استخدام وحدة `decimal` للحسابات المالية الدقيقة

عندما تتعامل مع أموال حقيقية أو حسابات تتطلب دقة تامة، استخدم وحدة `decimal` المُدمجة في Python:

````python
from decimal import Decimal

price = Decimal("19.99")  # لاحظ: كنص! وليس float مباشرة
quantity = Decimal("3")
total = price * quantity

print(total)  # 59.97  ← دقيق تمامًا!
````

> 🎯 **قاعدة ذهبية للمشاريع الحقيقية**: أي نظام يتعامل مع أموال فعلية (تطبيق متجر، نظام محاسبة، بنك) **يجب** أن يستخدم `Decimal` بدلاً من `float` العادي. هذا ليس اختياريًا - هو معيار صناعي معتمد عالميًا لتجنّب كوارث حسابية حقيقية حدثت في أنظمة مالية بسبب هذه المشكلة بالضبط.

### 🔄 مقارنة سريعة: متى تستخدم float ومتى decimal؟

| الحالة | الاستخدام الموصى به |
|---|---|
| حسابات علمية، إحداثيات، قياسات تقريبية | `float` ✅ |
| حسابات مالية، أسعار، فواتير، رواتب | `Decimal` ✅ |
| عرض سريع لرقم تقريبي للمستخدم | `round()` كافية |

## 5.6 التقريب بعمق: `round()` وتفاصيلها الدقيقة

### ✍️ الاستخدام الأساسي

````python
print(round(3.14159, 2))   # 3.14
print(round(3.14159, 0))   # 3.0
print(round(3.14159))      # 3  (بدون معامل ثانٍ، يُعيد int)
````

### ⚠️ سلوك غريب يجب معرفته: Banker's Rounding

````python
print(round(2.5))   # 2  ← ليس 3 كما قد تتوقع!
print(round(3.5))   # 4
print(round(0.5))   # 0  ← ليس 1!
````

هذا ليس خطأ - Python تستخدم أسلوبًا يُسمى **«التقريب المصرفي» (Banker's Rounding)**: عندما يكون الرقم بالضبط في المنتصف (مثل 0.5)، يُقرَّب لأقرب رقم **زوجي**. هذا الأسلوب مُستخدم إحصائيًا لتقليل التحيّز التراكمي في التقريب عبر آلاف العمليات.

> ⚠️ **تحذير عملي**: إذا كنت تتوقع التقريب «التقليدي» (0.5 دائمًا للأعلى)، ستُفاجأ بهذا السلوك. لحسابات حساسة، استخدم وحدة `decimal` مع إعدادات تقريب صريحة بدلاً من الاعتماد على `round()` الافتراضية.

## 5.7 دوال رياضية مُدمجة مفيدة

### 📐 وحدة `math`: دوال رياضية متقدمة

لدوال أكثر تخصصًا، نحتاج استيراد وحدة `math` المُدمجة:

````python
import math

print(math.sqrt(16))      # 4.0   ← الجذر التربيعي
print(math.ceil(4.1))     # 5     ← التقريب للأعلى دائمًا
print(math.floor(4.9))    # 4     ← التقريب للأسفل دائمًا
print(math.pi)             # 3.141592653589793
print(math.factorial(5))   # 120   ← المضروب (5!)
````

> 💡 **أول تعرّف على `import`**: هذا أول استخدام لنا لمفهوم **الاستيراد (import)** في الكتاب - طريقة Python لاستخدام كود إضافي مُنظَّم في «وحدات» (modules) منفصلة. سنتعمّق في هذا المفهوم بالتفصيل لاحقًا، لكن تذكّر الآن: أي وظيفة متقدمة غير متوفرة مباشرة، غالبًا موجودة في وحدة تحتاج استيرادها أولاً.

### 🔄 مقارنة: `round()` مقابل `math.floor()` مقابل `math.ceil()`

````python
number: float = 4.5

print(round(number))        # 4  (تقريب مصرفي للأقرب زوجي)
print(math.floor(number))   # 4  (دائمًا للأسفل)
print(math.ceil(number))    # 5  (دائمًا للأعلى)
````

## 5.8 تحويلات الأنواع الرقمية بعمق

### 🔄 int ↔ float

````python
# من int إلى float
x: int = 5
y: float = float(x)
print(y)  # 5.0

# من float إلى int (تحذير: يقطع الجزء العشري، لا يُقرِّب!)
a: float = 9.99
b: int = int(a)
print(b)  # 9  ← ليس 10! القطع وليس التقريب
````

> ⚠️ **تحذير حرج جدًا**: تحويل `float` إلى `int` باستخدام `int()` **يقطع** الجزء العشري ببساطة (Truncation)، ولا يُقرِّب أبدًا:

````python
print(int(9.99))   # 9  (ليس 10)
print(int(-9.99))  # -9 (ليس -10)
````

إذا أردت التقريب الفعلي قبل التحويل، استخدم `round()` أولاً:

````python
print(int(round(9.99)))  # 10  ← صحيح الآن
````

## 5.9 العمليات المختصرة (Augmented Assignment Operators)

### ✍️ الصيغة المختصرة لتحديث متغير

بدلاً من كتابة:

````python
counter: int = 10
counter = counter + 5
````

Python توفر صيغة مختصرة وأنيقة:

````python
counter: int = 10
counter += 5  # نفس تمامًا counter = counter + 5
print(counter)  # 15
````

### 📊 جدول العمليات المختصرة الكاملة

| الصيغة المختصرة | تُكافئ |
|---|---|
| `x += 5` | `x = x + 5` |
| `x -= 5` | `x = x - 5` |
| `x *= 5` | `x = x * 5` |
| `x /= 5` | `x = x / 5` |
| `x //= 5` | `x = x // 5` |
| `x %= 5` | `x = x % 5` |
| `x **= 5` | `x = x ** 5` |

````python
score: int = 0
score += 10  # score أصبحت 10
score += 5   # score أصبحت 15
score *= 2   # score أصبحت 30
print(score)  # 30
````

> ✅ **أفضل الممارسات**: استخدم الصيغ المختصرة دائمًا عند تحديث قيمة متغير بناءً على قيمته الحالية - أقصر، أوضح نية، وهي الأسلوب القياسي المتوقع من أي مبرمج Python محترف.

## 5.10 مقارنة الأرقام: عوامل المقارنة

سنتعمّق في استخدامها الكامل في الفصل السادس (التحكم بسير البرنامج)، لكن من الضروري التعرّف عليها الآن:

````python
print(5 > 3)    # True
print(5 < 3)    # False
print(5 == 5)   # True
print(5 != 3)   # True
print(5 >= 5)   # True
print(5 <= 4)   # False
````

### ⚠️ تحذير: مقارنة الأعداد العشرية مباشرة

````python
result: float = 0.1 + 0.2
print(result == 0.3)  # False!  ← بسبب فخ الدقة العشرية الذي شرحناه
````

**الحل الصحيح لمقارنة أعداد عشرية:**

````python
result: float = 0.1 + 0.2
is_close_enough: bool = abs(result - 0.3) < 0.0001
print(is_close_enough)  # True
````

> 💡 **نصيحة متقدمة**: في المشاريع الاحترافية، يُستخدم غالبًا `math.isclose()` الجاهزة بدل كتابة هذا المنطق يدويًا:

````python
import math
print(math.isclose(0.1 + 0.2, 0.3))  # True
````

## 5.12 الأعداد العشوائية: مقدمة سريعة لوحدة `random`

لن نتعمق هنا، لكن من المفيد التعرّف على هذه الأداة المُستخدَمة بكثرة لاحقًا في المشاريع (الألعاب، المحاكاة):

````python
import random

random_number: int = random.randint(1, 10)  # رقم عشوائي بين 1 و10 (شاملاً الطرفين)
print(random_number)

random_float: float = random.random()  # رقم عشوائي بين 0 و1
print(random_float)
````

## 5.13 أخطاء شائعة ونصائح احترافية

### ⚠️ الخطأ 1: القسمة على صفر

```python
result = 10 / 0  # ❌ ZeroDivisionError: division by zero
```

⚠️ **للقراءة فقط** — خطأ مقصود للتوضيح، لذا لا يُنفَّذ.

سنتعلّم كيف «نمسك» هذا الخطأ بأناقة في فصل معالجة الأخطاء (الفصل الحادي عشر). الآن، فقط تذكّر أن القسمة على صفر غير معرَّفة رياضيًا، وPython ترفضها بخطأ واضح بدل نتيجة غير منطقية.

### ⚠️ الخطأ 2: نسيان أن `/` تُعيد float دائمًا

````python
total_items: int = 10
groups: int = 3
items_per_group = total_items / groups  # 3.333... وليس 3

# إذا أردت عددًا صحيحًا من المجموعات:
items_per_group = total_items // groups  # 3  ← استخدم القسمة الصحيحة
````

### ⚠️ الخطأ 3: الثقة العمياء بنتائج float في المقارنات

````python
# كود خاطئ محتمل
if (0.1 + 0.2) == 0.3:
    print("متساويان")
else:
    print("غير متساويين")  # هذا ما سيُطبع فعليًا!
````

### 📋 المهمة

أنشئ ملفًا باسم `scientific_calculator.py` يبني حاسبة متكاملة من الطرفية:

**المتطلبات:**

3. تنفيذ العملية الصحيحة بناءً على اختيار المستخدم
4. معالجة حالة القسمة على صفر بشكل أنيق (رسالة واضحة بدل انهيار البرنامج - استخدام مبدئي بسيط، سنتعمّق في `try/except` لاحقًا)

### 💻 حل نموذجي كامل

<details>
<summary>✅ اضغط لرؤية الحل الكامل (حاول بنفسك أولاً!)</summary>

```python
# scientific_calculator.py
"""حاسبة علمية تفاعلية بسيطة من سطر الأوامر."""

print("=" * 40)
print("🧮  الحاسبة العلمية التفاعلية")
print("=" * 40)

# استقبال المدخلات
num1: float = float(input("أدخل الرقم الأول: "))
operator: str = input("أدخل العملية (+ - * / // % **): ").strip()
num2: float = float(input("أدخل الرقم الثاني: "))

# معالجة حالة خاصة: القسمة على صفر
if operator in ("/", "//", "%") and num2 == 0:
    print("\n❌ خطأ: لا يمكن القسمة على صفر!")
else:
    result: float = 0.0
    valid_operation: bool = True

    if operator == "+":
        result = num1 + num2
    elif operator == "-":
        result = num1 - num2
    elif operator == "*":
        result = num1 * num2
    elif operator == "/":
        result = num1 / num2
    elif operator == "//":
        result = num1 // num2
    elif operator == "%":
        result = num1 % num2
    elif operator == "**":
        result = num1 ** num2
    else:
        valid_operation = False
        print(f"\n❌ خطأ: العملية '{operator}' غير معروفة!")

    if valid_operation:
        print("\n" + "=" * 40)
        print(f"📊 النتيجة: {num1} {operator} {num2} = {round(result, 4)}")
        print("=" * 40)
```

⚠️ **للقراءة فقط** — خطأ مقصود للتوضيح، لذا لا يُنفَّذ.
