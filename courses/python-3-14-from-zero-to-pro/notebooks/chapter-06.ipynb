## 6.1 📖 مقدمة: من التنفيذ الخطي إلى اتخاذ القرار

كل البرامج التي كتبتها حتى الآن تشترك في صفة واحدة: **تنفيذ خطي (Linear Execution)** - تبدأ من السطر الأول، وتنتهي بالسطر الأخير، بدون أي انحراف. لكن البرامج الحقيقية ليست كذلك أبدًا. فكّر في أي تطبيق تستخدمه يوميًا: تطبيق البنك «يقرر» إظهار رسالة خطأ إن أدخلت كلمة مرور خاطئة، نظام الملاحة «يقرر» مسارًا بديلاً إن وُجد ازدحام. هذا «القرار» هو ما سنتعلمه الآن.

## 6.2 القيم المنطقية في صميم اتخاذ القرار

### 🔄 مراجعة سريعة: كل قرار يعتمد على `True` أو `False`

تذكر من الفصل الثالث: `bool` لها قيمتان فقط. كل جملة شرطية في Python تعتمد جوهريًا على تقييم تعبير ما إلى `True` أو `False`:

````python
age: int = 20
print(age >= 18)  # True  ← هذا هو أساس كل قرار برمجي
````

### 📊 عوامل المقارنة الكاملة (مراجعة وتوسيع)

| العامل | المعنى | مثال | النتيجة |
|---|---|---|---|
| `==` | يساوي | `5 == 5` | `True` |
| `!=` | لا يساوي | `5 != 3` | `True` |
| `>` | أكبر من | `5 > 3` | `True` |
| `<` | أصغر من | `5 < 3` | `False` |
| `>=` | أكبر أو يساوي | `5 >= 5` | `True` |
| `<=` | أصغر أو يساوي | `5 <= 4` | `False` |

## 6.4 التفريع: `if...else`

````python
age: int = 15

if age >= 18:
    print("مسموح لك بالدخول")
else:
    print("عذرًا، يجب أن تكون 18 سنة أو أكثر")
````

> 💡 **الفهم الجوهري**: `else` تعني «في أي حالة أخرى غير الشرط أعلاه» - لا تحتاج شرطًا خاصًا بها، فهي «كل ما تبقى».

## 6.5 شروط متعددة: `if...elif...else`

````python
score: int = 75

if score >= 90:
    grade: str = "ممتاز"
elif score >= 80:
    grade = "جيد جدًا"
elif score >= 70:
    grade = "جيد"
elif score >= 60:
    grade = "مقبول"
else:
    grade = "راسب"

print(f"تقديرك: {grade}")  # تقديرك: جيد
````

### 🔬 كيف تُقيَّم `elif` فعليًا؟ (فهم دقيق مهم)

Python تفحص الشروط **بالترتيب من الأعلى للأسفل**، وبمجرد أن يكون أحدها `True`، **تتوقف فورًا** عن فحص الباقي (حتى لو كانت شروط لاحقة صحيحة أيضًا):

````python
score: int = 95

if score >= 90:
    print("ممتاز")      # هذا يُطبع ويتوقف هنا
elif score >= 80:
    print("جيد جدًا")    # لن يُفحص حتى، رغم أن 95 >= 80 أيضًا صحيح!
````

> ⚠️ **خطأ شائع جدًا بسبب هذا المفهوم**: ترتيب الشروط **مهم جدًا**. إذا كتبت الشروط بترتيب خاطئ، ستحصل على نتائج غير منطقية:

```python
# ❌ ترتيب خاطئ - منطق معكوس
score: int = 95

if score >= 60:
    print("مقبول")   # هذا سيُطبع لـ 95! لأنه أول شرط صحيح يُقابَل
elif score >= 90:
    print("ممتاز")   # لن يصل هنا أبدًا لأي درجة أعلى من 60!
```

⚠️ **للقراءة فقط** — خطأ مقصود للتوضيح، لذا لا يُنفَّذ.

> ✅ **القاعدة الذهبية**: عند استخدام نطاقات متدرجة، رتّب الشروط دائمًا من **الأكثر تحديدًا (تشددًا) إلى الأقل**.

## 6.6 العوامل المنطقية: `and`, `or`, `not`

أحيانًا تحتاج لدمج أكثر من شرط واحد معًا.

### 🔗 `and`: كلا الشرطين يجب أن يكونا صحيحين

````python
age: int = 25
has_license: bool = True

if age >= 18 and has_license:
    print("يمكنك القيادة")
````

### 🔗 `or`: شرط واحد على الأقل يكفي

````python
is_weekend: bool = False
is_holiday: bool = True

if is_weekend or is_holiday:
    print("يوم إجازة!")
````

### 🔗 `not`: عكس القيمة المنطقية

````python
is_raining: bool = False

if not is_raining:
    print("يمكننا الخروج للنزهة")
````

### 📊 جدول الحقيقة المبسّط (Truth Table)

| A | B | `A and B` | `A or B` | `not A` |
|---|---|---|---|---|
| True | True | True | True | False |
| True | False | False | True | False |
| False | True | False | True | True |
| False | False | False | False | True |

### 🎯 دمج عوامل متعددة مع الأقواس للوضوح

````python
age: int = 25
has_ticket: bool = True
is_vip: bool = False

if (age >= 18 and has_ticket) or is_vip:
    print("مسموح بالدخول")
````

> ✅ **أفضل الممارسات**: عند دمج `and` و `or` معًا في نفس الشرط، استخدم الأقواس دائمًا لتوضيح الأولوية - حتى لو كانت النتيجة نفسها بدون أقواس، فالوضوح للقارئ أهم بكثير.

### ⚠️ تحذير: Short-Circuit Evaluation (التقييم القصير)

Python «تتوقف مبكرًا» عن تقييم الشرط إن أمكنها معرفة النتيجة النهائية مسبقًا:

````python
def expensive_check() -> bool:
    print("تم استدعاء الفحص المكلف!")
    return True

# مع and: إذا كان الشرط الأول False، لن يُستدعى الثاني إطلاقًا
if False and expensive_check():
    pass
# لن تُطبع "تم استدعاء الفحص المكلف!" أبدًا!

# مع or: إذا كان الشرط الأول True، لن يُستدعى الثاني إطلاقًا
if True or expensive_check():
    pass
# لن تُطبع "تم استدعاء الفحص المكلف!" أبدًا!
````

> 💡 **فائدة عملية مهمة**: هذا السلوك يُستخدم بذكاء لتجنّب أخطاء محتملة:

````python
user_data: dict = {}  # قاموس فارغ (سنتعلمه بالفصل التاسع)

if "name" in user_data and user_data["name"] != "":
    print("الاسم موجود وغير فارغ")
# إذا لم يكن "name" موجودًا أصلاً، and تتوقف عند الشرط الأول
# وتتجنب خطأ محاولة الوصول لمفتاح غير موجود
````

## 6.7 القيم «الصادقة» و«الكاذبة» ضمنيًا (Truthy & Falsy)

### 🎭 مفهوم فريد ومهم جدًا في Python

في Python، ليس فقط `True` و `False` يُستخدمان في الشروط - أي قيمة يمكن تقييمها ضمنيًا كـ«صادقة» (Truthy) أو «كاذبة» (Falsy):

````python
name: str = ""

if name:
    print("الاسم موجود")
else:
    print("الاسم فارغ")  # هذا سيُطبع! لأن النص الفارغ يُعتبر Falsy
````

### 📊 القيم التي تُعتبر Falsy (كل ما عداها يُعتبر Truthy)

| القيمة | النوع |
|---|---|
| `False` | bool |
| `0`, `0.0` | أرقام |
| `""` (نص فارغ) | str |
| `None` | NoneType |
| `[]` (قائمة فارغة) | list - سنتعلمها لاحقًا |
| `{}` (قاموس فارغ) | dict - سنتعلمه لاحقًا |

````python
value: int = 0

if value:
    print("له قيمة")
else:
    print("فارغ أو صفر")  # هذا سيُطبع
````

> ✅ **استخدام احترافي شائع جدًا**: التحقق من وجود قيمة مُدخلة من المستخدم:

```python
name: str = input("أدخل اسمك (اختياري): ").strip()

if name:
    print(f"مرحبًا {name}")
else:
    print("مرحبًا أيها الزائر المجهول")
```

🧾 **للقراءة فقط** — يحتاج إدخالًا يدويًا.

> ⚠️ **تحذير دقيق**: لا تخلط بين `if value:` (فحص Truthy/Falsy) و `if value == True:` (مقارنة صريحة). الأولى أكثر Pythonic (أسلوب Python الأصيل) والثانية أحيانًا تُسبب سلوكًا غير متوقع:

````python
value = 5
print(value == True)  # False!  لأن 5 لا يساوي True حرفيًا رغم أنه Truthy
print(bool(value))     # True   لأن 5 يُقيَّم كـ Truthy
````

## 6.8 الجُمل الشرطية المُدمجة: Ternary Expression

### ✍️ صيغة مختصرة لـ if/else في سطر واحد

````python
age: int = 20

# الطريقة التقليدية
if age >= 18:
    status = "بالغ"
else:
    status = "قاصر"

# الطريقة المختصرة (Ternary / Conditional Expression)
status = "بالغ" if age >= 18 else "قاصر"
````

كلاهما يُعطي نفس النتيجة بالضبط، لكن الصيغة الثانية أقصر وأحيانًا أوضح للحالات البسيطة.

> ✅ **متى تستخدم كل أسلوب؟**

| الحالة | الأسلوب المُفضَّل |
|---|---|
| إسناد قيمة بسيطة بناءً على شرط واحد | Ternary (مختصرة) |
| تنفيذ عدة أسطر أو منطق معقد | if/else التقليدية |
| أكثر من شرط (elif متعددة) | if/elif/else التقليدية فقط |

```python
# ✅ استخدام جيد للـ ternary - بسيط وواضح
discount = 0.2 if is_vip else 0.0

# ❌ استخدام سيء - معقد جدًا وغير مقروء كـ ternary
result = "A" if x > 90 else "B" if x > 80 else "C" if x > 70 else "D"
# في هذه الحالة، استخدم if/elif/else التقليدية بدلاً من هذا!
```

⚠️ **للقراءة فقط** — خطأ مقصود للتوضيح، لذا لا يُنفَّذ.

## 6.9 match-case: الأسلوب الحديث الثوري (Python 3.10+)

### 🆕 لماذا هذه الميزة مهمة جدًا؟

حتى إصدار Python 3.9، كانت اللغة تفتقر لأداة «تبديل» (Switch) الموجودة في لغات أخرى كثيرة (Java, C++, JavaScript). بدءًا من Python 3.10، ظهرت أداة `match-case` - لكنها أقوى بكثير من أي `switch` تقليدي، لأنها تدعم ما يُسمى **Pattern Matching** (مطابقة الأنماط).

### ✍️ الاستخدام الأساسي

````python
day: str = "Monday"

match day:
    case "Monday":
        print("بداية الأسبوع")
    case "Friday":
        print("نهاية أسبوع العمل")
    case "Saturday" | "Sunday":  # عامل | يعني "أو" هنا
        print("عطلة نهاية الأسبوع")
    case _:
        print("يوم عادي")
````

> 💡 **ملاحظة مهمة**: الرمز `_` في `case _:` يعني «أي قيمة أخرى» (مشابه لـ `else` في if/elif/else) - يجب أن يكون دائمًا **آخر** حالة في match.

### 🔄 متى تستخدم `match-case` بدلاً من `if/elif`؟

````python
# سيناريو مثالي لـ match-case: فحص قيمة واحدة ضد عدة احتمالات محددة
http_status: int = 404

match http_status:
    case 200:
        print("نجاح")
    case 404:
        print("غير موجود")
    case 500:
        print("خطأ في الخادم")
    case _:
        print("حالة غير معروفة")
````

### 🚀 قوة إضافية: مطابقة الأنماط المتقدمة (Pattern Matching الحقيقي)

هنا تكمن القوة الحقيقية لـ `match-case` - أبعد بكثير من مجرد مقارنة قيم بسيطة:

````python
# مطابقة أنواع البيانات والبنى (سنفهم هذا بعمق أكبر لاحقًا مع القوائم)
def describe_value(value):
    match value:
        case 0:
            return "صفر"
        case int() if value > 0:
            return "رقم صحيح موجب"
        case int() if value < 0:
            return "رقم صحيح سالب"
        case str():
            return "نص"
        case _:
            return "نوع غير معروف"

print(describe_value(5))       # رقم صحيح موجب
print(describe_value(-3))      # رقم صحيح سالب
print(describe_value("hi"))    # نص
````

> 🎯 لا تقلق إن بدت الأمثلة المتقدمة معقدة الآن - هذا طبيعي تمامًا. سنعود لـ `match-case` بعمق أكبر لاحقًا بعد تعلم القوائم والكائنات، حيث ستظهر قوتها الحقيقية بوضوح أكبر.

### 🔄 مقارنة شاملة: متى if/elif ومتى match-case؟

| الحالة | الأسلوب المُفضَّل |
|---|---|
| فحص نطاقات (مثل `score >= 90`) | `if/elif/else` ✅ |
| فحص قيمة واحدة ضد عدة قيم محددة | `match-case` ✅ |
| شروط منطقية معقدة (`and`/`or` متعددة) | `if/elif/else` ✅ |
| مطابقة أنواع بيانات أو بنى معقدة | `match-case` ✅ |

## 6.10 Walrus Operator `:=` - الأداة الذكية (Python 3.8+)

### 🤔 المشكلة التي يحلها

أحيانًا تحتاج لحساب قيمة واستخدامها في شرط **ونفس الوقت** تخزينها لاستخدام لاحق، مما يُسبب تكرارًا غير ضروري:

````python
# بدون walrus operator - تكرار الحساب مرتين
import random
number = random.randint(1, 10)

if number > 5:
    print(f"الرقم {number} أكبر من 5")
````

### ✍️ الحل بـ Walrus Operator

````python
import random

if (number := random.randint(1, 10)) > 5:
    print(f"الرقم {number} أكبر من 5")
````

العامل `:=` يُسند القيمة للمتغير **وفي نفس الوقت** يُعيد تلك القيمة لاستخدامها مباشرة في الشرط - كل هذا في سطر واحد.

> 💡 **متى تستخدمه فعليًا؟** هذا العامل اختياري بالكامل، ويُستخدم بكثرة أكثر في حلقات `while` (سنراها في الفصل القادم) وفي list comprehensions (الفصل الثامن). لا تُفرط باستخدامه في الشروط البسيطة إن كان يُقلّل الوضوح.

> ⚠️ **تحذير من الإفراط**: لا تستخدم `:=` فقط «لأنها أداة جديدة ومثيرة». استخدمها فقط عندما تُحسّن **الوضوح فعليًا** وتتجنب تكرار كود حقيقي.

## 6.11 الشروط المتداخلة (Nested Conditions)

````python
age: int = 25
has_id: bool = True

if age >= 18:
    if has_id:
        print("يمكنك الدخول")
    else:
        print("تحتاج إثبات هوية")
else:
    print("عمرك غير كافٍ")
````

> ⚠️ **تحذير احترافي**: الشروط المتداخلة بعمق (أكثر من 2-3 مستويات) تُصعّب قراءة الكود بشكل كبير. غالبًا يمكن تبسيطها بدمج الشروط بـ `and`:

```python
# بدلاً من التداخل العميق
if age >= 18:
    if has_id:
        if not is_banned:
            print("مسموح")

# الأفضل: دمج الشروط
if age >= 18 and has_id and not is_banned:
    print("مسموح")
```

🧾 **للقراءة فقط** — مثال مقتطف يعتمد على `is_banned` من مثال سابق.

### ✅ تقنية احترافية: Early Return / Guard Clauses

تقنية متقدمة (سنتعمق فيها أكثر مع الدوال في الفصل العاشر) تُجنّبك التداخل العميق:

````python
# بدلاً من هذا (متداخل)
def check_access(age, has_id):
    if age >= 18:
        if has_id:
            return "مسموح"
        else:
            return "تحتاج هوية"
    else:
        return "عمر غير كافٍ"

# استخدم هذا (أوضح - Guard Clauses)
def check_access(age, has_id):
    if age < 18:
        return "عمر غير كافٍ"
    if not has_id:
        return "تحتاج هوية"
    return "مسموح"
````

> 💡 لا تقلق إن لم تفهم صيغة الدالة بالكامل بعد - النقطة المهمة هنا هي **مبدأ** تقليل التداخل، وسنعود لهذا بعمق في فصل الدوال.

## 6.13 جدول مرجعي شامل للفصل

| الأداة | الاستخدام | مثال |
|---|---|---|
| `if` | شرط بسيط | `if x > 5:` |
| `if/else` | شرط مع بديل | `if x > 5: ... else: ...` |
| `if/elif/else` | شروط متعددة | `if...elif...elif...else` |
| `and` | كلا الشرطين صحيح | `x > 5 and x < 10` |
| `or` | أحد الشرطين على الأقل | `x < 5 or x > 10` |
| `not` | عكس القيمة | `not is_valid` |
| Ternary | شرط مختصر بسطر واحد | `"A" if x else "B"` |
| `match-case` | مطابقة قيمة لعدة احتمالات | `match x: case 1: ...` |
| `:=` | إسناد واستخدام في آن واحد | `if (n := get()) > 5:` |

### 📋 المهمة

**المتطلبات:**
1. طلب كلمة مرور من المستخدم
2. التحقق من الشروط التالية باستخدام methods النصوص من الفصل الرابع:
   - الطول 8 أحرف على الأقل
   - تحتوي رقمًا واحدًا على الأقل
   - تحتوي حرفًا كبيرًا واحدًا على الأقل (إن كانت إنجليزية)
   - تحتوي رمزًا خاصًا واحدًا على الأقل (مثل `!@#$%`)
3. حساب «نقاط القوة» بناءً على عدد الشروط المُستوفاة
4. تصنيف القوة: ضعيفة (0-1 شروط) / متوسطة (2-3 شروط) / قوية (4 شروط)

### 💻 حل نموذجي كامل

<details>
<summary>✅ اضغط لرؤية الحل الكامل (حاول بنفسك أولاً!)</summary>

```python
# password_checker.py
"""نظام تقييم قوة كلمة المرور."""

SPECIAL_CHARS: str = "!@#$%^&*()_+-="

password: str = input("أدخل كلمة مرور لتقييمها: ")

# فحص كل شرط على حدة
has_min_length: bool = len(password) >= 8
has_digit: bool = any(char.isdigit() for char in password)
has_upper: bool = any(char.isupper() for char in password)
has_special: bool = any(char in SPECIAL_CHARS for char in password)

# حساب عدد الشروط المُستوفاة
strength_score: int = sum([has_min_length, has_digit, has_upper, has_special])

# تحديد مستوى القوة
if strength_score <= 1:
    strength_level: str = "ضعيفة ❌"
elif strength_score <= 3:
    strength_level = "متوسطة ⚠️"
else:
    strength_level = "قوية ✅"

# طباعة التقرير
print("\n" + "=" * 40)
print("🔐  تقرير تقييم كلمة المرور")
print("=" * 40)
print(f"الطول (8+ أحرف): {'✅' if has_min_length else '❌'}")
print(f"يحتوي رقمًا: {'✅' if has_digit else '❌'}")
print(f"يحتوي حرفًا كبيرًا: {'✅' if has_upper else '❌'}")
print(f"يحتوي رمزًا خاصًا: {'✅' if has_special else '❌'}")
print("-" * 40)
print(f"النتيجة: {strength_score}/4")
print(f"مستوى القوة: {strength_level}")
print("=" * 40)
```

**ملاحظة على الحل**: استخدمنا هنا `any()` مع تعبير يُشبه list comprehension (`char.isdigit() for char in password`) - هذا تشويق بسيط لمفاهيم ستتعلمها بعمق أكبر في الفصل الثامن (القوائم). لا تقلق إن بدت غامضة الآن؛ لاحظ فقط كيف تتكامل المفاهيم مع تقدمك في الكتاب.

## 6.17 📌 ملخص الفصل (Cheat Sheet)

```python
# الشرط الأساسي
if condition:
    pass
elif other_condition:
    pass
else:
    pass

# العوامل المنطقية
x and y   # كلاهما True
x or y    # أحدهما True
not x     # عكس x

# Ternary Expression
result = "A" if condition else "B"

# match-case (Python 3.10+)
match value:
    case 1:
        pass
    case 2 | 3:  # أو
        pass
    case _:  # الحالة الافتراضية
        pass

# Walrus Operator
if (n := get_value()) > 5:
    print(n)

# Truthy / Falsy
if my_string:  # True إن لم يكن فارغًا
    pass
```

🧾 **للقراءة فقط** — مثال مقتطف يعتمد على `condition`، `get_value`، `my_string` من مثال سابق.
