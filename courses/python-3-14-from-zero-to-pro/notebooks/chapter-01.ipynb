## 1.1 📖 مقدمة: لماذا هذا الفصل قبل أي سطر كود؟

## 1.2 ما هي البرمجة فعليًا؟

لنبدأ بتعريف بسيط وصادق:

> **البرمجة = إعطاء تعليمات دقيقة جدًا لجهاز «غبي» حرفيًا، لكنه سريع جدًا في التنفيذ.**

الكلمة المفتاحية هنا هي **«دقيقة»**. الكمبيوتر لا يفهم السياق، ولا يخمّن قصدك، ولا يقول «أعتقد أنك تقصد كذا». هو ينفّذ **بالضبط** ما كتبته، لا أكثر ولا أقل.

### 🍳 تشبيه وصفة الطبخ

تخيل أنك تكتب وصفة لشخص لم يدخل مطبخًا في حياته، ولا يملك أي حس سليم، لكنه ينفذ التعليمات بدقة روبوتية:

**وصفة سيئة (غامضة):**

```
اصنع بيضة مقلية
```

الشخص الحرفي لا يعرف: هل يكسر البيضة أولاً؟ هل يشغّل الموقد قبل أم بعد وضع المقلاة؟ كم من الوقت يقليها؟

**وصفة جيدة (دقيقة كالكود):**

```
1. أحضر مقلاة من الخزانة
2. ضع المقلاة على الموقد
3. شغّل الموقد على حرارة متوسطة
4. أحضر بيضة واحدة من الثلاجة
5. اكسر البيضة في المقلاة
6. انتظر 3 دقائق
7. أطفئ الموقد
8. انقل البيضة إلى طبق
```

هذا بالضبط ما تفعله عندما تكتب كودًا. أنت لا «تطلب» من الكمبيوتر شيئًا، أنت **تصف له كل خطوة** بترتيب منطقي صارم.

## 1.3 الخوارزمية: الوصفة قبل الكود

الكلمة التقنية لـ«الوصفة» هذه هي **Algorithm (خوارزمية)**.

> **الخوارزمية** = سلسلة محددة ومنتهية من الخطوات لحل مشكلة معينة.

لاحظ ثلاث خصائص مهمة في هذا التعريف:

| الخاصية | المعنى | مثال مخالف (خطأ) |

| **منتهية** | لها بداية ونهاية واضحتان | حلقة لا تتوقف أبدًا ❌ |
| **قابلة للتنفيذ** | يمكن فعلاً القيام بها | «فكّر بعمق في الحل» ❌ |

## 1.4 كيف «يفهم» الكمبيوتر الكود الذي تكتبه؟

هذا سؤال يتجنبه كثير من الكتب، لكن فهمه يزيل غموضًا كبيرًا من ذهن المبتدئ.

الكمبيوتر في الحقيقة **لا يفهم** Python إطلاقًا. المعالج (CPU) يفهم لغة واحدة فقط: **اللغة الثنائية** (0 و 1)، تسمى **Machine Code**.

السؤال إذن: كيف يتحول الكود الذي نكتبه بكلمات شبه إنجليزية (`print`, `if`, `for`) إلى لغة الأصفار والآحاد؟

### 🔄 مترجم (Compiler) مقابل مفسّر (Interpreter)

هناك طريقتان رئيسيتان لحل هذه المشكلة في عالم البرمجة:

**الطريقة الأولى - الترجمة الكاملة مسبقًا (Compiled Languages)**
مثل لغة C++. يأخذ برنامج يسمى **Compiler** كل الكود دفعة واحدة، ويترجمه بالكامل إلى ملف تنفيذي بلغة الآلة، **قبل** أي تشغيل فعلي. تشبّه هذه العملية بترجمة كتاب كامل إلى لغة أخرى قبل طباعته.

**الطريقة الثانية - الترجمة الفورية أثناء التشغيل (Interpreted Languages)**
وهذه طريقة Python. برنامج يسمى **Interpreter** يقرأ الكود **سطرًا بسطر**، ويترجم وينفّذ كل سطر فورًا قبل الانتقال للسطر التالي. يشبه هذا مترجمًا فوريًا يترجم كلام متحدث مباشرة أثناء إلقائه الخطاب.

### 🔄 جدول مقارنة: لماذا هذا يهمك عمليًا؟

| الجانب | لغات مُفسَّرة (Python) | لغات مُترجَمة (C++, Rust) |
|---|---|---|
| **السرعة في التطوير** | ✅ سريعة جدًا - اكتب وشغّل فورًا | ❌ أبطأ - يجب الترجمة أولاً |
| **سرعة التنفيذ** | ❌ أبطأ نسبيًا | ✅ أسرع بكثير |
| **اكتشاف الأخطاء** | عند الوصول للسطر الخاطئ فقط | ✅ قبل التشغيل بالكامل |
| **الأفضل لـ** | التعلم، السكربتات، النماذج الأولية | الأنظمة الحساسة للأداء |

> 💡 **لماذا هذا مهم لك كمبتدئ؟**
> لأن Python ستسامحك! إن كان لديك خطأ في السطر 50، فستُنفَّذ الأسطر من 1 إلى 49 بنجاح تام قبل أن تتوقف عند الخطأ. هذا رائع للتعلم التدريجي، لكنه يعني أيضًا أن اختبار الكود بشكل كامل قبل الاعتماد عليه أمر ضروري (سنتعلم هذا في فصل الاختبارات لاحقًا).

## 1.6 عقلية حل المشكلات: المهارة الحقيقية

إليك سرًا يغفل عنه كثيرون: **90% من البرمجة الاحترافية ليست كتابة كود، بل التفكير قبل الكتابة.**

### 🧩 منهجية حل أي مشكلة برمجية (ستستخدمها طوال الكتاب)

عندما تواجه أي مشكلة، مهما بدت بسيطة أو معقدة، اتبع هذه الخطوات دائمًا:

```
1. افهم المشكلة تمامًا
   → ما هي المُدخلات (Inputs)؟
   → ما هي المُخرجات المطلوبة (Outputs)؟

2. فكّر بأمثلة يدوية بسيطة
   → جرّب حل المشكلة بيدك على ورقة بمثال صغير

3. اكتب الخوارزمية (Pseudocode)
   → بخطوات واضحة بلغتك الخاصة، بدون قلق من الـ syntax

4. حوّل الخوارزمية إلى كود حقيقي
   → هذه أسهل خطوة فعليًا إن أتقنت ما قبلها!

5. اختبر الكود بحالات متعددة
   → حالة عادية، حالة حدّية (edge case)، حالة خاطئة
```

### 🎯 مثال تطبيقي كامل على المنهجية

لنطبق هذه الخطوات على مشكلة واقعية:

> **المشكلة:** تريد برنامجًا يخبرك إن كانت سنة معينة «سنة كبيسة» (366 يومًا) أم لا.

**الخطوة 1: فهم المشكلة**
- المُدخل: رقم سنة (مثل 2024)
- المُخرج: نعم أو لا

**الخطوة 2: جرّب بأمثلة يدوية**
- هل تعرف قاعدة السنة الكبيسة؟ (قابلة للقسمة على 4، لكن ليس على 100 إلا إذا كانت قابلة للقسمة على 400)
- 2024 ÷ 4 = 506 بدون باقي → كبيسة ✅
- 1900 ÷ 4 = بدون باقي، لكن ÷ 100 أيضًا بدون باقي، و÷400 بها باقي → **ليست** كبيسة ❌
- 2000 ÷ 400 = بدون باقي → كبيسة ✅

**الخطوة 3: الخوارزمية (Pseudocode)**

```
1. خذ السنة كمُدخل
2. إذا كانت السنة قابلة للقسمة على 400 → كبيسة
3. وإلا، إذا كانت قابلة للقسمة على 100 → ليست كبيسة
4. وإلا، إذا كانت قابلة للقسمة على 4 → كبيسة
5. وإلا → ليست كبيسة
```

**الخطوة 4: الكود الحقيقي (نظرة تشويقية فقط - سنتعلم هذا بالتفصيل لاحقًا)**

````python
def is_leap_year(year: int) -> bool:
    """تحديد ما إذا كانت السنة كبيسة أم لا."""
    if year % 400 == 0:
        return True
    elif year % 100 == 0:
        return False
    elif year % 4 == 0:
        return True
    else:
        return False

# اختبار الدالة
print(is_leap_year(2024))  # True
print(is_leap_year(1900))  # False
print(is_leap_year(2000))  # True
````

**الخطوة 5: اختبار بحالات متعددة**
- حالة عادية: 2023 → ليست كبيسة ✅
- حالة حدّية: 2000 (قرن لكن قابل للقسمة على 400) ✅
- حالة حدّية أخرى: 1900 (قرن غير قابل للقسمة على 400) ✅

> 🎉 **لاحظ شيئًا مهمًا جدًا**: لم تفهم كل سطر في الكود أعلاه، وهذا طبيعي تمامًا! الهدف من هذا المثال ليس تعليمك `def` أو `%` الآن، بل إثبات أن **المنهجية** (الخطوات 1-3) هي الجزء الصعب والمهم فعليًا، وأن كتابة الكود النهائي (الخطوة 4) تصبح سهلة نسبيًا بمجرد أن تفكر بوضوح.

## 1.7 أخطاء ذهنية شائعة عند المبتدئين (تجنّبها من اليوم الأول)

### ⚠️ الخطأ 1: «يجب أن أحفظ كل شيء عن ظهر قلب»

**الحقيقة**: حتى المبرمجون المحترفون ذوو خبرة 20 عامًا يبحثون في جوجل يوميًا. المهارة الحقيقية هي **فهم المفاهيم**، لا حفظ الأوامر. الحفظ يأتي تلقائيًا بالممارسة المتكررة.

### ⚠️ الخطأ 2: «إذا ظهر لي خطأ (Error)، فأنا فاشل»

**الحقيقة**: الأخطاء (Errors) هي **أفضل أصدقائك** في رحلة التعلم. كل رسالة خطأ هي معلم يشرح لك بالضبط ماذا حدث. سنتعلم في الفصل القادم كيف «نقرأ» رسائل الأخطاء بدل الخوف منها.

### ⚠️ الخطأ 3: «يجب أن أفهم كل التفاصيل قبل الانتقال للموضوع التالي»

**الحقيقة**: التعلم البرمجي **حلزوني** وليس خطيًا. ستتعلم مفهومًا بسيطًا، تستخدمه بشكل سطحي، ثم تعود له لاحقًا بفهم أعمق بعد أن ترى تطبيقات أكثر. لا تعلّق نفسك في فصل واحد لأسابيع.

### ⚠️ الخطأ 4: «النسخ من مصادر أخرى حرام أو غش»

**الحقيقة**: كل المبرمجين المحترفين يبحثون ويرجعون لمصادر (التوثيق الرسمي، Stack Overflow). الفرق بين المبتدئ والمحترف ليس «عدم البحث»، بل **فهم ما تنسخه** قبل استخدامه.

## 1.8 بناء «عضلة» حل المشكلات - تمارين عملية

الآن حان دورك. هذه التمارين **لا تحتاج كتابة كود** - فقط ورقة وقلم، أو حتى تفكير ذهني. الهدف تدريب عقلك على التفكير الخوارزمي قبل الانتقال للكود الفعلي في الفصول القادمة.

### 🟢 تمرين 1 (مستوى مبتدئ)

### 🟡 تمرين 2 (مستوى متوسط)

## 1.9 مشروع تطبيقي صغير للفصل: «مصمم الخوارزميات»

بما أننا لم نكتب كودًا بعد، مشروع هذا الفصل مختلف عن باقي الفصول: هو **تمرين تفكير منهجي كامل**.

### 📋 المهمة

اختر **موقفًا واقعيًا** من حياتك اليومية (مثال: «كيف أختار ماذا ألبس حسب حالة الطقس»، أو «كيف أرتب أولويات مهامي اليومية»)، ثم:

1. حدد **المُدخلات** بدقة (ما المعلومات التي تحتاجها لاتخاذ القرار؟)
2. حدد **المُخرج** المطلوب (ما القرار النهائي؟)
3. اكتب **خوارزمية كاملة** بخطوات مرقّمة وواضحة تمامًا
4. اختبر خوارزميتك ذهنيًا بـ 3 حالات مختلفة (تأكد أنها تعمل منطقيًا في كل حالة)

> 🎯 **لماذا هذا المشروع مهم؟** لأنه يحاكي **بالضبط** ما ستفعله في كل مشروع برمجي طوال حياتك المهنية: تحويل مشكلة واقعية غامضة إلى خطوات منطقية دقيقة، **قبل** لمس لوحة المفاتيح لكتابة كود فعلي. المبرمجون الذين يتخطون هذه الخطوة هم من يقضون ساعات في «تصحيح» كود مبني على تفكير غير واضح من الأساس.

## 1.11 📌 ملخص الفصل (Cheat Sheet)

| المفهوم | التعريف المختصر |
|---|---|
| **Algorithm (خوارزمية)** | سلسلة محددة ومنتهية من الخطوات لحل مشكلة |
| **Pseudocode** | وصف خطوات الحل بلغة بشرية بسيطة قبل كتابة الكود الفعلي |
| **Interpreter (مُفسِّر)** | برنامج يترجم وينفّذ الكود سطرًا بسطر أثناء التشغيل (طريقة Python) |
| **Compiler (مُترجِم)** | برنامج يترجم كل الكود دفعة واحدة قبل أي تشغيل |
| **Decomposition (تفكيك)** | تقسيم مشكلة كبيرة معقدة إلى خطوات صغيرة وبسيطة |
| **Edge Case (حالة حدّية)** | حالة استثنائية أو نادرة يجب اختبار الكود ضدها |
