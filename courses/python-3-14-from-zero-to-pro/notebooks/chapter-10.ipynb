## 10.1 لماذا تُعتبر الدوال نقطة تحوّل في رحلتك البرمجية؟

تخيّل أنك تكتب برنامجاً لحساب مساحة دائرة في خمسة أماكن مختلفة من الكود. إذا احتجت لاحقاً تعديل طريقة الحساب (مثلاً لتحسين الدقة)، ستحتاج للبحث والتعديل في خمسة أماكن! وإذا نسيت مكاناً واحداً، سيظهر خطأ غامض يصعب تتبعه.

هذا بالضبط ما تحله **الدوال (Functions)**: بدلاً من تكرار نفس الكود، تكتبه **مرة واحدة** بداخل دالة، ثم «تستدعيها» (تطلب تنفيذها) من أي مكان تحتاجه، بقدر ما تشاء من المرات.

````python
# بدون دالة - تكرار مزعج وخطير
radius1 = 5
area1 = 3.14159 * radius1 ** 2
print(f"المساحة: {area1}")

radius2 = 8
area2 = 3.14159 * radius2 ** 2
print(f"المساحة: {area2}")

# مع دالة - اكتب مرة، استخدم دائماً
def calculate_circle_area(radius: float) -> float:
    return 3.14159 * radius ** 2

print(f"المساحة: {calculate_circle_area(5)}")
print(f"المساحة: {calculate_circle_area(8)}")
````

الدوال ليست مجرد «اختصار لتقليل الكتابة» — إنها أداة تفكير. المبرمج المحترف ينظر لأي مشكلة كبيرة ويسأل: «كيف أقسّمها إلى قطع أصغر، كل قطعة تقوم بمهمة واحدة واضحة؟» هذا المبدأ يُسمى **Separation of Concerns** (فصل الاهتمامات)، وهو حجر الأساس لكل كود احترافي، صغيراً كان أو ضخماً بحجم أنظمة شركات عملاقة.

بنهاية هذا الفصل، ستكتب دوالاً احترافية، تفهم كيف تتدفق البيانات داخل وخارج الدالة، وتتقن أحد أهم المفاهيم في Python: **النطاق (Scope)**.

## 10.2 أساسيات تعريف واستدعاء الدالة

### الصيغة الأساسية

```python
def function_name(parameters) -> return_type:
    """Docstring: وصف مختصر لما تفعله الدالة."""
    # جسم الدالة
    return value
```

🧾 **للقراءة فقط** — مثال مقتطف يعتمد على `return_type`، `value` من مثال سابق.

````python
def greet(name: str) -> str:
    """تُرجع رسالة ترحيب للمستخدم."""
    return f"أهلاً بك، {name}!"

message = greet("سارة")
print(message)  # أهلاً بك، سارة!
````

**تشريح الدالة:**

| الجزء | الوصف |
|---|---|
| `def` | كلمة مفتاحية تُعلن بداية تعريف دالة |
| `greet` | اسم الدالة (نفس قواعد تسمية المتغيرات: snake_case) |
| `(name: str)` | المعاملات (Parameters) مع Type Hints |
| `-> str` | Type Hint لنوع القيمة المُرجعة |
| `"""..."""` | Docstring - توثيق وظيفة الدالة |
| `return` | القيمة التي تُعاد للمستدعي |

> 💡 **نصيحة الخبراء**: اختر أسماء دوال تبدأ بفعل يصف ما تفعله: `calculate_total()`, `is_valid()`, `get_user_data()`, `send_email()`. اسم الدالة الجيد يجعل الكود «يُقرأ كجملة إنجليزية» دون الحاجة لتعليقات إضافية.

### Docstrings: التوثيق الاحترافي

````python
def calculate_bmi(weight_kg: float, height_m: float) -> float:
    """
    حساب مؤشر كتلة الجسم (BMI).

    المعاملات:
        weight_kg: الوزن بالكيلوغرام
        height_m: الطول بالمتر

    تُرجع:
        قيمة BMI كرقم عشري
    """
    return weight_kg / (height_m ** 2)

# يمكنك قراءة التوثيق في أي وقت باستخدام:
print(calculate_bmi.__doc__)
help(calculate_bmi)
````

### دالة بدون `return`

````python
def print_separator() -> None:
    """تطبع خطاً فاصلاً. لا تُرجع أي قيمة."""
    print("-" * 40)

result = print_separator()
print(result)  # None
````

> ⚠️ **تحذير شائع**: إذا لم تكتب `return` صراحة، الدالة تُرجع `None` تلقائياً. هذا سبب شائع لخطأ "لماذا متغيري أصبح `None`؟« — المبرمج يستدعي دالة تطبع نتيجة لكن ينسى أنها لا »تُرجع" شيئاً فعلياً.

````python
# ⚠️ مثال يُظهر خطأً مقصودًا: نلتقطه هنا ليكتمل «تشغيل الكل»،
# وسترى نوع الخطأ ورسالته في المخرجات كما في الطرفية.
try:
    # خطأ شائع
    def add(a: int, b: int) -> None:
        print(a + b)  # يطبع فقط، لا يُرجع!

    result = add(3, 5)  # result = None وليس 8!
    print(result + 1)  # 💥 TypeError: unsupported operand type(s)
except (Exception, SystemExit) as _err:
    print(f"⚠️ {type(_err).__name__}: {_err}")
````

````python
# الصح
def add(a: int, b: int) -> int:
    return a + b

result = add(3, 5)  # result = 8
print(result + 1)  # 9
````

### Parameters مقابل Arguments

مصطلحان كثيراً ما يُخلط بينهما:
- **Parameter (معامل)**: الاسم المُعرَّف في تعريف الدالة
- **Argument (وسيط)**: القيمة الفعلية المُمرَّرة عند الاستدعاء

````python
def greet(name: str) -> str:  # name = parameter
    return f"Hello, {name}"

greet("Ahmed")  # "Ahmed" = argument
````

### المعاملات الموضعية والمسماة (Positional & Keyword Arguments)

````python
def create_profile(name: str, age: int, city: str) -> str:
    return f"{name}, {age} سنة، من {city}"

# موضعية (Positional) - الترتيب مهم جداً
print(create_profile("Ahmed", 25, "Riyadh"))

# مسماة (Keyword) - الترتيب لا يهم
print(create_profile(city="Riyadh", name="Ahmed", age=25))

# مزيج - الموضعية أولاً، ثم المسماة
print(create_profile("Ahmed", city="Riyadh", age=25))
````

### القيم الافتراضية (Default Parameters)

````python
def greet(name: str, greeting: str = "أهلاً") -> str:
    return f"{greeting}, {name}!"

print(greet("سارة"))              # أهلاً, سارة!
print(greet("عمر", "مرحباً"))      # مرحباً, عمر!
print(greet("لينا", greeting="صباح الخير"))  # صباح الخير, لينا!
````

> ⚠️ **تحذير حرج - فخ القيم الافتراضية القابلة للتعديل (Mutable Default Argument)**

هذا من أخطر وأشهر الأخطاء في Python، حتى بين المبرمجين ذوي الخبرة:

```python
# ❌ خطأ خطير جداً
def add_item(item: str, cart: list = []) -> list:
    cart.append(item)
    return cart

cart1 = add_item("apple")
print(cart1)  # ['apple']

cart2 = add_item("banana")
print(cart2)  # ['apple', 'banana']  💥 لماذا تذكرت العنصر السابق؟!
```

⚠️ **للقراءة فقط** — خطأ مقصود للتوضيح، لذا لا يُنفَّذ.

**السبب**: القيمة الافتراضية `[]` تُنشأ **مرة واحدة فقط** عند تعريف الدالة، وليس في كل استدعاء! لذا كل الاستدعاءات التي لا تمرر `cart` تشترك في **نفس القائمة بالذاكرة**.

````python
# ✅ الحل الصحيح: استخدم None كقيمة افتراضية
def add_item(item: str, cart: list | None = None) -> list:
    if cart is None:
        cart = []  # قائمة جديدة في كل استدعاء
    cart.append(item)
    return cart

cart1 = add_item("apple")
print(cart1)  # ['apple']

cart2 = add_item("banana")
print(cart2)  # ['banana']  ✅ صحيح الآن!
````

> 🔄 **مقارنة**: هذا السلوك مرتبط بكون Python تُنشئ كائنات القيم الافتراضية مرة واحدة عند **تعريف** الدالة (وقت التحميل)، وليس عند **استدعائها**. لغات أخرى مثل JavaScript لا تملك هذا الفخ لأن طريقة التعامل مع المعاملات الافتراضية تختلف جوهرياً. **القاعدة الذهبية في Python**: لا تستخدم أبداً `list`, `dict`, أو `set` كقيمة افتراضية مباشرة — استخدم `None` دائماً كبديل آمن.

### `*args`: عدد غير محدد من المعاملات الموضعية

````python
def calculate_sum(*numbers: float) -> float:
    """تجمع أي عدد من الأرقام الممررة."""
    total = 0
    for num in numbers:
        total += num
    return total

print(calculate_sum(1, 2, 3))           # 6
print(calculate_sum(1, 2, 3, 4, 5))     # 15
print(calculate_sum())                   # 0

# داخل الدالة، numbers تصبح tuple
def show_type(*args):
    print(type(args))  # <class 'tuple'>

show_type(1, 2, 3)
````

### `**kwargs`: عدد غير محدد من المعاملات المسماة

````python
def create_user(**details: str) -> dict:
    """تنشئ قاموس مستخدم من أي عدد من الحقول المسماة."""
    return details

user = create_user(name="Ahmed", age=25, city="Jeddah")
print(user)
# {'name': 'Ahmed', 'age': 25, 'city': 'Jeddah'}

# داخل الدالة، kwargs تصبح dict
def show_type(**kwargs):
    print(type(kwargs))  # <class 'dict'>

show_type(a=1, b=2)
````

### ترتيب المعاملات الكامل

````python
def full_example(
    positional: str,
    default: int = 10,
    *args: int,
    keyword_only: str = "default_value",
    **kwargs: str
) -> None:
    print(f"positional: {positional}")
    print(f"default: {default}")
    print(f"args: {args}")
    print(f"keyword_only: {keyword_only}")
    print(f"kwargs: {kwargs}")

full_example("A", 20, 1, 2, 3, keyword_only="X", extra="Y")
````

**الناتج:**

```
positional: A
default: 20
args: (1, 2, 3)
keyword_only: X
kwargs: {'extra': 'Y'}
```

🧾 **للقراءة فقط** — مثال مقتطف يعتمد على `A`، `X` من مثال سابق.

| الترتيب الإلزامي | الوصف |
|---|---|
| 1. Positional parameters | معاملات عادية بدون/مع قيمة افتراضية |
| 2. `*args` | معاملات موضعية إضافية غير محددة العدد |
| 3. Keyword-only parameters | يجب تمريرها بالاسم (بعد `*args` أو `*`) |
| 4. `**kwargs` | معاملات مسماة إضافية غير محددة العدد |

### Keyword-Only و Positional-Only Arguments (Python 3.8+)

````python
# ⚠️ مثال يُظهر خطأً مقصودًا: نلتقطه هنا ليكتمل «تشغيل الكل»،
# وسترى نوع الخطأ ورسالته في المخرجات كما في الطرفية.
try:
    # استخدام * لإجبار ما بعدها أن تكون keyword-only
    def create_order(product: str, *, quantity: int = 1, urgent: bool = False) -> str:
        return f"{quantity}x {product} {'(عاجل)' if urgent else ''}"

    # print(create_order("Laptop", 2))  
    # 💥 TypeError: يجب تمرير quantity بالاسم

    print(create_order("Laptop", quantity=2, urgent=True))  # ✅ صحيح

    # استخدام / لإجبار ما قبلها أن تكون positional-only
    def divide(a: float, b: float, /) -> float:
        return a / b

    print(divide(10, 2))      # ✅ صحيح
    # print(divide(a=10, b=2))  
    # 💥 TypeError: a و b يجب تمريرهما بالموضع فقط
except (Exception, SystemExit) as _err:
    print(f"⚠️ {type(_err).__name__}: {_err}")
````

> 💡 **نصيحة الخبراء**: استخدم `*` لإجبار الوضوح في المعاملات التي قد يُساء فهمها (مثل `urgent=True` بدلاً من `True` غامضة)، واستخدم `/` عندما يكون اسم المعامل تفصيلاً داخلياً لا يهم المستخدم (شائع في مكتبات built-in مثل `len(obj, /)`).

## 10.4 القيمة المُرجعة (Return Values) بعمق

### إرجاع قيم متعددة

````python
def get_min_max(numbers: list[float]) -> tuple[float, float]:
    """تُرجع أصغر وأكبر قيمة في القائمة."""
    return min(numbers), max(numbers)

low, high = get_min_max([5, 2, 9, 1, 7])
print(f"الأصغر: {low}, الأكبر: {high}")  # الأصغر: 1, الأكبر: 9
````

في الحقيقة، Python تُرجع **tuple واحدة** هنا (`(1, 9)`) ثم يحدث **Unpacking** تلقائي إلى المتغيرين.

### الخروج المبكر (Early Return / Guard Clauses)

تعلمنا هذا المفهوم في الفصل السادس؛ الدوال هي المكان المثالي لتطبيقه:

````python
def calculate_discount(price: float, is_member: bool) -> float:
    """حساب السعر بعد الخصم."""
    if price <= 0:
        return 0.0  # خروج مبكر - لا داعي لمتابعة الحساب

    if not is_member:
        return price  # خروج مبكر - لا خصم

    return price * 0.85  # الكود الرئيسي، بدون تداخل if متعدد
````

> 🔄 **مقارنة**: بدون Guard Clauses، سيبدو الكود هكذا مع تداخل مزعج:

````python
def calculate_discount_messy(price: float, is_member: bool) -> float:
    if price > 0:
        if is_member:
            return price * 0.85
        else:
            return price
    else:
        return 0.0
````

النمط الأول أسهل بكثير للقراءة والصيانة، خاصة مع تعقّد الشروط لاحقاً.

### دوال بدون return صريح مقابل `return None`

````python
def log_message(message: str) -> None:
    """تسجل رسالة. لا تُرجع شيئاً (None صراحة)."""
    print(f"[LOG]: {message}")
    return None  # اختياري لكنه أوضح من حيث القصد

def validate_age(age: int) -> bool | None:
    """تتحقق من صحة العمر؛ تُرجع None إذا كانت القيمة غير منطقية."""
    if age < 0 or age > 150:
        return None
    return age >= 18
````

## 10.5 النطاق (Scope): أين «تعيش» المتغيرات؟

هذا من أهم المفاهيم التي يسيء المبتدئون فهمها. **النطاق** يحدد أين يمكن الوصول لمتغير معين في الكود.

### Local Scope و Global Scope

````python
# ⚠️ مثال يُظهر خطأً مقصودًا: نلتقطه هنا ليكتمل «تشغيل الكل»،
# وسترى نوع الخطأ ورسالته في المخرجات كما في الطرفية.
try:
    x = 10  # متغير عام (Global)

    def my_function():
        y = 20  # متغير محلي (Local) - يعيش فقط داخل الدالة
        print(x)  # يمكن قراءة المتغير العام من داخل الدالة
        print(y)

    my_function()
    print(x)  # 10 - متاح
    # print(y)  
    # 💥 NameError: name 'y' is not defined -- y لا توجد خارج الدالة!
except (Exception, SystemExit) as _err:
    print(f"⚠️ {type(_err).__name__}: {_err}")
````

> ⚠️ **تحذير شائع**: المبتدئون يستغربون لماذا متغير عرّفوه داخل دالة «يختفي» خارجها. هذا ليس خطأ — إنه التصميم المقصود للنطاق، ويمنع الفوضى في البرامج الكبيرة.

### لماذا لا يمكن تعديل متغير عام من داخل دالة مباشرة؟

````python
# ⚠️ مثال يُظهر خطأً مقصودًا: نلتقطه هنا ليكتمل «تشغيل الكل»،
# وسترى نوع الخطأ ورسالته في المخرجات كما في الطرفية.
try:
    counter = 0

    def increment():
        counter += 1  # 💥 UnboundLocalError!

    increment()
except (Exception, SystemExit) as _err:
    print(f"⚠️ {type(_err).__name__}: {_err}")
````

**لماذا يحدث هذا الخطأ؟** لأن Python عندما ترى `counter += 1` داخل الدالة، تفترض أن `counter` متغير **محلي** جديد (لأنك تحاول الكتابة إليه)، لكنك تحاول قراءته قبل تعريفه محلياً، فتحدث مشكلة.

### الكلمة المفتاحية `global`

````python
counter = 0

def increment():
    global counter  # أخبر Python: استخدم المتغير العام، لا تنشئ واحداً محلياً
    counter += 1

increment()
increment()
print(counter)  # 2
````

> ⚠️ **تحذير احترافي**: استخدام `global` يُعتبر من أفضل الممارسات **السيئة** (Anti-pattern) في البرمجة الحديثة! لأنه يجعل الدوال «تعتمد» على حالة خارجية خفية، مما يصعّب تتبع الأخطاء وفهم الكود. **البديل الأفضل دائماً**:

````python
# ✅ الطريقة الاحترافية: مرر القيمة وأعد النتيجة
def increment(counter: int) -> int:
    return counter + 1

counter = 0
counter = increment(counter)
counter = increment(counter)
print(counter)  # 2
````

> 🔄 **مقارنة**: في البرمجة الكائنية (الفصل 13)، سنرى حلاً أنظف بكثير لهذه المشكلة باستخدام `self.counter` داخل الكائنات (Objects)، مما يلغي الحاجة لـ `global` نهائياً في الكود الاحترافي.

### قاعدة LEGB

Python تبحث عن أي متغير بالترتيب التالي:

| المستوى | الاسم | الوصف |
|---|---|---|
| 1 | **L**ocal | داخل الدالة الحالية |
| 2 | **E**nclosing | داخل أي دالة «محيطة» (دوال متداخلة - الفصل 12) |
| 3 | **G**lobal | على مستوى الملف كله |
| 4 | **B**uilt-in | الأسماء المدمجة في Python مثل `len`, `print` |

````python
x = "global x"

def outer():
    x = "enclosing x"

    def inner():
        x = "local x"
        print(x)  # local x - توجد في المستوى المحلي، تتوقف هنا

    inner()
    print(x)  # enclosing x

outer()
print(x)  # global x
````

> 💡 **نصيحة الخبراء**: كلما قلّت اعتمادية دالتك على متغيرات خارجية (Global)، وكلما اعتمدت فقط على معاملاتها (Parameters)، أصبحت أسهل للاختبار، وأسهل لإعادة الاستخدام، وأقل عرضة للأخطاء الغامضة. هذا المبدأ يُسمى **Pure Functions** (الدوال النقية): الدالة التي تُعطي نفس النتيجة دائماً لنفس المدخلات، ولا تُغيّر أي شيء خارج نطاقها.

````python
# دالة نقية (Pure Function) - يُفضّل هذا النمط
def add_tax(price: float, tax_rate: float) -> float:
    return price * (1 + tax_rate)

# دالة غير نقية - تعتمد على وتُعدّل حالة خارجية
total = 0

def add_to_total(amount: float) -> None:
    global total
    total += amount
````

## 10.6 تمرير المعاملات: القيمة مقابل المرجع

هذا المفهوم يُربك الكثير من المبرمجين القادمين من لغات أخرى.

````python
def modify_number(n: int) -> None:
    n += 10
    print(f"داخل الدالة: {n}")

x = 5
modify_number(x)
print(f"خارج الدالة: {x}")

# الناتج:
# داخل الدالة: 15
# خارج الدالة: 5   -- لم يتغير!
````

لكن مع القوائم (كائنات قابلة للتعديل):

````python
def modify_list(lst: list) -> None:
    lst.append(100)
    print(f"داخل الدالة: {lst}")

my_list = [1, 2, 3]
modify_list(my_list)
print(f"خارج الدالة: {my_list}")

# الناتج:
# داخل الدالة: [1, 2, 3, 100]
# خارج الدالة: [1, 2, 3, 100]  -- تغيّرت فعلاً!
````

**السبب**: Python تُمرر دائماً «مرجع» (Reference) للكائن، لكن التأثير يختلف حسب نوع الكائن:
- الأنواع **غير القابلة للتعديل** (int, float, str, tuple): أي عملية كـ `n += 10` تُنشئ كائناً **جديداً** بالكامل، فالمتغير الأصلي خارج الدالة لا يتأثر.
- الأنواع **القابلة للتعديل** (list, dict, set): العمليات مثل `.append()` تُعدّل **نفس الكائن** في الذاكرة، فالتغيير يظهر خارج الدالة أيضاً.

> ⚠️ **تحذير حرج**: هذا السلوك مصدر شائع جداً للأخطاء الغامضة. إذا مررت قائمة لدالة ولم تقصد تعديلها، انسخها أولاً:

````python
def process_safely(lst: list) -> list:
    lst_copy = lst.copy()  # اعمل على نسخة
    lst_copy.append(999)
    return lst_copy
````

> 🔄 **مقارنة مع لغات أخرى**: هذا المفهوم يُسمى أحياناً "Pass by Object Reference"، وهو مختلف عن "Pass by Value" الصِرف (كما في C عند تمرير أنواع بسيطة) و"Pass by Reference" الصِرف (كما في بعض حالات ++C). Python تقف في منتصف الطريق: المرجع نفسه يُمرر بالقيمة، لكن الكائن المُشار إليه يبقى نفسه.

## 10.7 دوال من الدرجة الأولى (First-Class Functions) - مقدمة

في Python، الدوال هي «مواطنون من الدرجة الأولى» (First-Class Citizens) — يمكن التعامل معها كأي قيمة أخرى: تخزينها في متغير، تمريرها كمعامل لدالة أخرى، أو إرجاعها من دالة.

````python
def square(x: int) -> int:
    return x ** 2

# تخزين الدالة في متغير (بدون أقواس استدعاء!)
my_func = square
print(my_func(5))  # 25

# تمرير دالة كمعامل لدالة أخرى
def apply_function(func, value: int) -> int:
    return func(value)

print(apply_function(square, 6))  # 36
````

مثال عملي باستخدام `sorted()` مع معامل `key` (دالة):

````python
students = [
    {"name": "Ahmed", "score": 85},
    {"name": "Sara", "score": 92},
    {"name": "Omar", "score": 78}
]

def get_score(student: dict) -> int:
    return student["score"]

sorted_students = sorted(students, key=get_score, reverse=True)
for s in sorted_students:
    print(f"{s['name']}: {s['score']}")

# الناتج:
# Sara: 92
# Ahmed: 85
# Omar: 78
````

## 10.8 Type Hints المتقدمة في الدوال

````python
# استخدام | للسماح بأكثر من نوع (Python 3.10+)
def format_value(value: int | float | str) -> str:
    return f"القيمة: {value}"

# استخدام Optional (أو | None) لمعامل قد يكون غائباً
def find_user(user_id: int) -> dict | None:
    users = {1: {"name": "Ahmed"}, 2: {"name": "Sara"}}
    return users.get(user_id)

# Callable: Type Hint لمعامل يكون دالة
from collections.abc import Callable

def apply_twice(func: Callable[[int], int], value: int) -> int:
    return func(func(value))

def add_five(x: int) -> int:
    return x + 5

print(apply_twice(add_five, 10))  # 20
````

## 10.9 مشروع تطبيقي: حاسبة علمية بالدوال

سنبني حاسبة علمية مُنظّمة بالكامل بدوال، تدعم العمليات الأساسية والمتقدمة، مع معالجة آمنة للأخطاء الشائعة.

```python
import math

def add(a: float, b: float) -> float:
    """جمع رقمين."""
    return a + b

def subtract(a: float, b: float) -> float:
    """طرح رقمين."""
    return a - b

def multiply(a: float, b: float) -> float:
    """ضرب رقمين."""
    return a * b

def divide(a: float, b: float) -> float | None:
    """قسمة رقمين، مع حماية من القسمة على صفر."""
    if b == 0:
        print("❌ خطأ: لا يمكن القسمة على صفر.")
        return None
    return a / b

def power(base: float, exponent: float) -> float:
    """رفع رقم لأس معين."""
    return base ** exponent

def square_root(number: float) -> float | None:
    """الجذر التربيعي، مع حماية من الأرقام السالبة."""
    if number < 0:
        print("❌ خطأ: لا يمكن حساب الجذر التربيعي لرقم سالب.")
        return None
    return math.sqrt(number)

def percentage(part: float, whole: float) -> float | None:
    """حساب النسبة المئوية لـ part من whole."""
    if whole == 0:
        print("❌ خطأ: القيمة الكلية لا يمكن أن تكون صفراً.")
        return None
    return (part / whole) * 100

# --- قاموس يربط الرمز بالدالة المقابلة (نمط احترافي) ---
OPERATIONS: dict[str, tuple] = {
    "+": (add, 2),
    "-": (subtract, 2),
    "*": (multiply, 2),
    "/": (divide, 2),
    "^": (power, 2),
    "sqrt": (square_root, 1),
    "%": (percentage, 2),
}

def display_menu() -> None:
    """طباعة قائمة العمليات المتاحة."""
    print("\n" + "=" * 40)
    print("🧮 الحاسبة العلمية")
    print("=" * 40)
    print("العمليات المتاحة:")
    print("  +     : جمع")
    print("  -     : طرح")
    print("  *     : ضرب")
    print("  /     : قسمة")
    print("  ^     : أس (قوة)")
    print("  sqrt  : جذر تربيعي")
    print("  %     : نسبة مئوية")
    print("  q     : خروج")
    print("=" * 40)

def run_calculator() -> None:
    """تشغيل الحاسبة في حلقة تفاعلية."""
    display_menu()

    while True:
        choice = input("\nاختر العملية: ").strip()

        if choice == "q":

            break

        if choice not in OPERATIONS:
            print("❌ عملية غير معروفة. حاول مجدداً.")

        func, num_args = OPERATIONS[choice]

            else:
                x = float(input("أدخل الرقم الأول: "))
                y = float(input("أدخل الرقم الثاني: "))
                result = func(x, y)

            if result is not None:

        except ValueError:
            print("❌ خطأ: الرجاء إدخال أرقام صحيحة فقط.")

if __name__ == "__main__":
    run_calculator()
```

### 🟡 مستوى متوسط

**تمرين 4**: اكتب دالة `is_valid_password(password: str) -> tuple[bool, str]` تتحقق أن كلمة المرور: طولها 8 أحرف على الأقل، تحتوي رقماً واحداً على الأقل، وحرفاً كبيراً واحداً على الأقل. تُرجع `(True, "صالحة")` أو `(False, "سبب الرفض")`.

**تمرين 5**: اكتب دالة `filter_products(products: list[dict], **criteria) -> list[dict]` تُرجع المنتجات المطابقة لأي معايير مُمرَّرة (مثلاً `category="إلكترونيات"`).

<details>
<summary>✅ الحل</summary>

````python
# تمرين 4
def is_valid_password(password: str) -> tuple[bool, str]:
    if len(password) < 8:
        return False, "يجب أن تكون 8 أحرف على الأقل"
    if not any(char.isdigit() for char in password):
        return False, "يجب أن تحتوي على رقم واحد على الأقل"
    if not any(char.isupper() for char in password):
        return False, "يجب أن تحتوي على حرف كبير واحد على الأقل"
    return True, "صالحة"

valid, message = is_valid_password("abc123")
print(f"{valid}: {message}")  # False: يجب أن تكون 8 أحرف على الأقل

valid, message = is_valid_password("Abcd1234")
print(f"{valid}: {message}")  # True: صالحة

# تمرين 5
def filter_products(products: list[dict], **criteria) -> list[dict]:
    result = []
    for product in products:
        if all(product.get(key) == value for key, value in criteria.items()):
            result.append(product)
    return result

products = [
    {"name": "لابتوب", "category": "إلكترونيات", "price": 3500},
    {"name": "كرسي", "category": "أثاث", "price": 450},
    {"name": "ماوس", "category": "إلكترونيات", "price": 50},
]

electronics = filter_products(products, category="إلكترونيات")
print(electronics)
# [{'name': 'لابتوب', ...}, {'name': 'ماوس', ...}]
````

### 🔴 تحدي

**تمرين 6**: اكتب دالة `memoize(func)` تأخذ دالة أخرى كمعامل، وتُرجع نسخة «محسّنة» منها تُخزّن (تتذكر) نتائج الاستدعاءات السابقة بنفس المعاملات، بدلاً من إعادة حسابها (هذا تمهيد لمفهوم Decorators في الفصل 12).

<details>
<summary>✅ الحل</summary>

````python
def memoize(func):
    """تُرجع نسخة من func تخزّن النتائج السابقة."""
    cache: dict = {}

    def wrapper(*args):
        if args in cache:
            print(f"(من الذاكرة المؤقتة لـ {args})")
            return cache[args]
        result = func(*args)
        cache[args] = result
        return result

    return wrapper

def slow_square(n: int) -> int:
    import time
    time.sleep(1)  # محاكاة عملية بطيئة
    return n ** 2

fast_square = memoize(slow_square)

print(fast_square(5))  # يأخذ ثانية واحدة، النتيجة: 25
print(fast_square(5))  # فوري! (من الذاكرة المؤقتة لـ (5,))
print(fast_square(7))  # يأخذ ثانية واحدة، النتيجة: 49
````

> 💡 ملاحظة: مكتبة Python القياسية توفر هذا جاهزاً عبر `functools.lru_cache` — سنتعرف عليها في الفصل 12 عند دراسة Decorators بعمق.
</details>

## 10.12 ملخص الفصل (Cheat Sheet)

```python
# تعريف دالة أساسية
def func_name(param: type) -> return_type:
    """Docstring."""
    return value

# قيم افتراضية
def func(a: int, b: int = 10) -> int: ...

# ⚠️ لا تستخدم mutable كقيمة افتراضية
def bad(lst: list = []): ...            # ❌
def good(lst: list | None = None):      # ✅
    if lst is None:
        lst = []

# *args و **kwargs
def func(*args, **kwargs): ...
func(1, 2, 3, key="value")

# Keyword-only و Positional-only
def func(a, b, /, c, d, *, e, f): ...
#         ^positional-only  ^keyword-only

# إرجاع قيم متعددة (tuple unpacking)
def minmax(nums: list) -> tuple:
    return min(nums), max(nums)
low, high = minmax([1, 5, 3])

# Scope
x = 10                   # Global
def f():
    y = 20                # Local
    global x; x = 99       # تعديل Global (تجنّبه قدر الإمكان)

# الدوال كقيم (First-Class Functions)
my_func = some_function   # بدون ()
sorted(data, key=my_func) # تمرير كمعامل
```

⚠️ **للقراءة فقط** — خطأ مقصود للتوضيح، لذا لا يُنفَّذ.

| المفهوم | القاعدة الذهبية |
|---|---|
| القيم الافتراضية | لا تستخدم list/dict/set مباشرة، استخدم None |
| التعديل على الحالة الخارجية | تجنّب global، مرر وأعد القيم بدلاً منه |
| وضوح الاستدعاء | استخدم Keyword Arguments مع أكثر من 3 معاملات |
| عدد غير محدد من المدخلات | `*args` للموضعية، `**kwargs` للمسماة |
| الدوال النقية (Pure Functions) | يُفضّل دائماً: نفس المدخل = نفس المخرج، بدون آثار جانبية |
