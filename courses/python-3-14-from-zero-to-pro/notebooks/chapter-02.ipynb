## 2.4 أداة uv: إدارة حديثة وفائقة السرعة

### 🔄 لماذا uv بدلاً من pip التقليدي؟

لسنوات، كان `pip` هو الأداة الوحيدة لتثبيت مكتبات Python الإضافية. لكن في عامي 2023-2024، ظهرت أداة ثورية اسمها **uv** (من شركة Astral، صانعة أداة Ruff الشهيرة) غيّرت قواعد اللعبة بالكامل.

| الخاصية | pip (التقليدية) | uv (الحديثة) |
|---|---|---|
| **السرعة** | بطيئة نسبيًا | ⚡ أسرع بـ 10-100 مرة |
| **إدارة البيئات الافتراضية** | تحتاج أداة منفصلة (venv) | ✅ مدمجة بالكامل |
| **إدارة إصدارات Python** | غير مدعومة | ✅ مدعومة مباشرة |
| **حل التعارضات بين الحزم** | بدائي | ✅ ذكي ومتقدم |
| **الاعتماد العالمي 2025** | يتراجع تدريجيًا | 📈 يتزايد بسرعة |

> 💡 **قرار تعليمي مهم**: سنعتمد في هذا الكتاب بالكامل على `uv` بدلاً من `pip` التقليدية، لأنها أصبحت المعيار الحديث الموصى به من قبل خبراء المجتمع، وستوفر عليك أخطاء ومشاكل كثيرة كان يعاني منها المبتدئون سابقًا مع الأدوات القديمة.

### 📥 تثبيت uv

**على Windows (PowerShell):**

```powershell
powershell -ExecutionPolicy ByPass -c "irm https://astral.sh/uv/install.ps1 | iex"
```

🧾 **للقراءة فقط** — أوامر `powershell` تُنفَّذ من الطرفية، لا من الخلية.

**على macOS / Linux:**

```bash
curl -LsSf https://astral.sh/uv/install.sh | sh
```

🧾 **للقراءة فقط** — أوامر `bash` تُنفَّذ من الطرفية، لا من الخلية.

**التحقق من التثبيت (على أي نظام):**

أغلق الطرفية وافتح واحدة جديدة، ثم:

```bash
uv --version
```

🧾 **للقراءة فقط** — أوامر `bash` تُنفَّذ من الطرفية، لا من الخلية.

يجب أن تظهر نتيجة مثل:

```
uv 0.5.x
```

> ✅ **ميزة إضافية رائعة**: بإمكان `uv` تثبيت إصدارات Python نفسها! مثلاً، يمكنك كتابة `uv python install 3.14` وستُثبَّت تلقائيًا دون الحاجة للذهاب لموقع Python الرسمي يدويًا. سنستخدم هذه الميزة بكثرة لاحقًا.

## 2.6 أول تواصل مع Python: الـ REPL

### 💬 ما هو REPL؟

قبل أن نكتب ملفات كود حقيقية، دعنا نتعرّف على طريقة تفاعلية رائعة لتجربة Python فورًا: الـ **REPL** (اختصار لـ Read-Eval-Print-Loop).

فكّر في REPL كـ«محادثة مباشرة» مع Python: تكتب سطرًا، يُنفَّذ فورًا، وترى النتيجة مباشرة.

### 🎮 جرّب بنفسك

افتح الطرفية (Terminal في VS Code، أو cmd/Terminal في نظامك)، واكتب:

```bash
python
```

🧾 **للقراءة فقط** — أوامر `bash` تُنفَّذ من الطرفية، لا من الخلية.

(أو `python3` على Mac/Linux إن لزم الأمر)

ستظهر لك واجهة مثل هذه:

```
Python 3.14.0 (main, ...) 
Type "help", "copyright", "credits" or "license" for more information.
>>>
```

🧾 **للقراءة فقط** — مقتطف غير مكتمل.

الرمز `>>>` يعني أن Python **جاهزة وتنتظر** أوامرك. جرّب كتابة:

```python
>>> 2 + 2
4
```

🧾 **للقراءة فقط** — جلسة طرفية تفاعلية.

```python
>>> "مرحبا" + " " + "بالعالم"
'مرحبا بالعالم'
```

🧾 **للقراءة فقط** — جلسة طرفية تفاعلية.

```python
>>> print("أنا أتعلم Python!")
أنا أتعلم Python!
```

🧾 **للقراءة فقط** — جلسة طرفية تفاعلية.

> 💡 **نصيحة عملية**: استخدم REPL دائمًا لتجربة سطر كود سريع أو اختبار فكرة صغيرة، بدل إنشاء ملف كامل لكل تجربة. أغلب المبرمجين المحترفين يستخدمونه يوميًا لاختبارات سريعة.

للخروج من REPL في أي وقت:

```python
>>> exit()
```

🧾 **للقراءة فقط** — جلسة طرفية تفاعلية.

### ✍️ الخطوة 3: إنشاء أول ملف

في VS Code، أنشئ ملفًا جديدًا باسم `hello.py`. لكن بدلاً من مجرد `Hello World` التقليدي المُعاد استخدامه آلاف المرات، سنكتب شيئًا تفاعليًا بسيطًا يعطيك إحساسًا حقيقيًا بقوة البرمجة من اليوم الأول:

```python

age_in_days = int(age) * 365
print(f"هذا يعني أنك عشت تقريبًا {age_in_days} يومًا على هذه الأرض!")
```

🧾 **للقراءة فقط** — يحتاج إدخالًا يدويًا.

## 2.8 فن قراءة رسائل الخطأ (مهارة تُغيّر رحلتك بالكامل)

### 🎯 لماذا هذا القسم هو الأهم في الفصل؟

تذكّر ما قلناه في الفصل الأول: الأخطاء ليست أعداءك، بل معلّموك. لكن لتستفيد منها، يجب أن تتعلّم **كيف تقرأها**. دعنا نرتكب خطأ مقصودًا ونحلّله معًا.

### 🔬 تجربة: خطأ مقصود

عدّل ملف `hello.py` وأضف هذا السطر الخاطئ عمدًا في النهاية:

```python
print("النتيجة النهائية: " + age_in_days)
```

🧾 **للقراءة فقط** — مثال مقتطف يعتمد على `age_in_days` من مثال سابق.

شغّل الملف وستظهر رسالة خطأ تشبه هذه:

```
Traceback (most recent call last):
  File "hello.py", line 10, in <module>
    print("النتيجة النهائية: " + age_in_days)
                                  ~~~~~~~~~~~~~~~~~~~~~~~^~~~~~~~~~~~~~
TypeError: can only concatenate str (not "int") to str
```

⚠️ **للقراءة فقط** — ناتج خطأ متوقّع.

### 🧩 تشريح رسالة الخطأ خطوة بخطوة

هذه الرسالة تبدو مرعبة لأول وهلة، لكنها في الحقيقة **خريطة دقيقة** تخبرك بكل شيء تحتاجه:

```
Traceback (most recent call last):
  ↑ هذا فقط يعني "إليك تتبّع مسار الخطأ"

File "hello.py", line 10, in <module>
  ↑ اسم الملف ورقم السطر بالضبط حيث حدثت المشكلة

print("النتيجة النهائية: " + age_in_days)
  ↑ السطر الفعلي الذي سبّب الخطأ (معروض لك مباشرة!)

TypeError: can only concatenate str (not "int") to str
  ↑ نوع الخطأ + شرح دقيق: "لا يمكن دمج str مع int"
```

⚠️ **للقراءة فقط** — ناتج خطأ متوقّع.

### 🗝️ القاعدة الذهبية لقراءة أي خطأ في Python

```
1. اقرأ السطر الأخير أولاً (نوع الخطأ + الشرح) - هذا 80% من الحل
2. انظر لرقم السطر المذكور - اذهب إليه مباشرة في كودك
3. اسأل نفسك: "ماذا يحاول هذا السطر فعله، ولماذا يرفضه Python؟"
```

في مثالنا: `age_in_days` هو رقم (int)، بينما النص قبله هو str. Python لا تسمح بدمج (concatenate) نوعين مختلفين بـ `+` مباشرة. الحل:

```python
print("النتيجة النهائية: " + str(age_in_days))
```

🧾 **للقراءة فقط** — مثال مقتطف يعتمد على `age_in_days` من مثال سابق.

> 💡 **لا تقلق إن لم تفهم `str` و `int` بعمق الآن** - سنشرحهما بالتفصيل الكامل في الفصل القادم. الهدف هنا فقط تعلّم **منهجية** قراءة الخطأ، لا حل هذا الخطأ تحديدًا.

### 📚 أنواع الأخطاء الأكثر شيوعًا للمبتدئين (مرجع سريع)

| نوع الخطأ | متى يظهر | مثال شائع |
|---|---|---|
| `SyntaxError` | خطأ في قواعد كتابة الكود نفسه | نسيان `:` بعد `if` |
| `NameError` | استخدام متغير غير موجود | خطأ إملائي في اسم متغير |
| `TypeError` | خلط أنواع بيانات غير متوافقة | جمع نص مع رقم مباشرة |
| `IndentationError` | مشكلة في المسافات البادئة | عدم محاذاة الأسطر بشكل صحيح |
| `ZeroDivisionError` | قسمة أي رقم على صفر | `10 / 0` |

> ⚠️ **تحذير من عادة سيئة شائعة جدًا**: كثير من المبتدئين عند رؤية خطأ طويل، يصابون بالذعر وينسخون **الرسالة بأكملها** إلى جوجل أو ChatGPT دون قراءتها أولاً. تدرّب على قراءة السطر الأخير بنفسك أولاً - ستندهش كم مرة ستفهم المشكلة بنفسك خلال ثوانٍ فقط.

## 2.9 لمحة عن البيئات الافتراضية (Virtual Environments)

### 🤔 لماذا نحتاج هذا المفهوم من البداية؟

تخيل أنك تعمل على مشروعين مختلفين: الأول يحتاج نسخة قديمة من مكتبة معينة، والثاني يحتاج أحدث نسخة من نفس المكتبة. إذا ثبّت كل شيء «عالميًا» على جهازك، ستحدث تعارضات حتمًا.

**البيئة الافتراضية (Virtual Environment)** هي حل بسيط وعبقري: «فقاعة» معزولة خاصة بكل مشروع، تحتوي نسخة Python والمكتبات الخاصة به فقط، دون التأثير على بقية مشاريعك أو نظامك.

> 💡 سنتعمّق في هذا المفهوم بالتفصيل الكامل في فصل لاحق مخصص لإدارة المشاريع الاحترافية. الآن، فقط تذكّر: **كل مشروع حقيقي يحتاج بيئته الافتراضية الخاصة** - هذه قاعدة ذهبية لا تُخالف أبدًا في الممارسة الاحترافية.

### 🎮 تجربة سريعة بـ uv

داخل مجلد مشروعك، جرّب:

```bash
# إنشاء بيئة افتراضية جديدة
uv venv

# تفعيلها (الأوامر تختلف حسب نظامك)
# على Windows:
.venv\Scripts\activate

# على Mac/Linux:
source .venv/bin/activate
```

🧾 **للقراءة فقط** — أوامر `bash` تُنفَّذ من الطرفية، لا من الخلية.

بعد التفعيل، ستلاحظ تغيّرًا بسيطًا في الطرفية (عادة اسم البيئة يظهر بين قوسين)، وهذا يعني أنك الآن «داخل» بيئة معزولة. سنستخدم هذا بكثرة بدءًا من مشاريعنا الأولى في الفصول القادمة.

## 2.10 أخطاء التهيئة الشائعة وحلولها السريعة

### ⚠️ المشكلة 1: `'python' is not recognized as an internal or external command`

**السبب**: لم يتم تفعيل خيار "Add to PATH" أثناء التثبيت (على Windows).

**الحل**: أعد تثبيت Python، وتأكد من تفعيل ذلك الخيار هذه المرة، أو أضف المسار يدويًا لمتغيرات البيئة (خطوة متقدمة، البحث عنها بـ"Add Python to PATH manually" يعطيك حلولاً مفصلة).

### ⚠️ المشكلة 2: VS Code لا «يرى» Python المُثبَّت

**السبب**: VS Code لم يكتشف تلقائيًا أين Python مُثبَّت.

**الحل**: اضغط `Ctrl+Shift+P` (أو `Cmd+Shift+P`)، اكتب `Python: Select Interpreter`، واختر الإصدار الصحيح من القائمة الظاهرة.

### ⚠️ المشكلة 3: الأحرف العربية تظهر كرموز غريبة

**السبب**: مشكلة في ترميز الملف (Encoding).

**الحل**: تأكد في VS Code (أسفل يمين الشاشة) أن الترميز هو `UTF-8`. إذا لم يكن كذلك، اضغط عليه واختر `Reopen with Encoding` ثم `UTF-8`.
