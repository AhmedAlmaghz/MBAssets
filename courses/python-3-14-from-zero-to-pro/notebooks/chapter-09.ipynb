## 9.2 القواميس (Dictionaries)

### ما هو القاموس؟

القاموس هو مجموعة من أزواج **«مفتاح: قيمة»** (Key: Value). تماماً كقاموس اللغة الحقيقي: تبحث عن **كلمة** (المفتاح) لتجد **معناها** (القيمة).

````python
# قاموس حقيقي
# "Serendipity" -> "اكتشاف شيء جميل بالصدفة"

ages: dict[str, int] = {
    "Sara": 25,
    "Omar": 30,
    "Lina": 28
}

print(ages["Sara"])  # 25
````

### خصائص القاموس الجوهرية

| الخاصية | الوصف |
|---|---|
| **غير مرتب منطقياً** لكنه **يحفظ ترتيب الإدخال** | منذ Python 3.7، القواميس تحافظ على ترتيب إضافة العناصر (لكن لا تعتمد عليه منطقياً كترتيب «أول/آخر» بمعنى الفرز) |
| **قابل للتعديل (Mutable)** | يمكنك إضافة/حذف/تعديل العناصر بعد الإنشاء |
| **المفاتيح فريدة** | لا يمكن تكرار نفس المفتاح؛ أي قيمة جديدة لمفتاح موجود تستبدل القديمة |
| **المفاتيح يجب أن تكون Immutable** | str, int, float, tuple مسموحة كمفاتيح؛ list أو dict ممنوعة |
| **القيم يمكن أن تكون أي شيء** | نصوص، أرقام، قوائم، حتى قواميس أخرى |

### إنشاء القواميس

````python
# الطريقة الأولى: الأقواس المعقوفة
empty_dict: dict = {}
person: dict[str, str] = {"name": "Khaled", "city": "Riyadh"}

# الطريقة الثانية: دالة dict()
person2 = dict(name="Noor", city="Cairo")  # بدون علامات تنصيص للمفاتيح هنا!

# الطريقة الثالثة: من قائمة أزواج (tuples)
pairs = [("a", 1), ("b", 2)]
person3 = dict(pairs)
print(person3)  # {'a': 1, 'b': 2}

# قاموس متداخل (Nested Dictionary) - شائع جداً في البيانات الحقيقية
company: dict[str, dict] = {
    "employee_1": {"name": "Sami", "salary": 8000},
    "employee_2": {"name": "Huda", "salary": 9500}
}
print(company["employee_1"]["name"])  # Sami
````

> 💡 **نصيحة الخبراء**: استخدم `dict[str, int]` كـ Type Hint عندما تعرف أنواع المفاتيح والقيم. إذا كانت القيم متنوعة الأنواع (كما في مثال `student` أعلاه)، استخدم `dict[str, object]` أو الأفضل: استخدم **Dataclass** (سنتعلمها لاحقاً في الفصل 13) بدلاً من قاموس بقيم مختلطة.

### الوصول للقيم والتعامل مع المفاتيح المفقودة

````python
# ⚠️ مثال يُظهر خطأً مقصودًا: نلتقطه هنا ليكتمل «تشغيل الكل»،
# وسترى نوع الخطأ ورسالته في المخرجات كما في الطرفية.
try:
    student = {"name": "Ahmed", "age": 22}

    # الطريقة الأولى: الأقواس المربعة - تُطلق خطأ إذا المفتاح غير موجود
    print(student["name"])  # Ahmed
    # print(student["grade"])  # KeyError: 'grade'  💥

    # الطريقة الثانية: get() - آمنة، تُعيد None إذا لم يوجد المفتاح
    print(student.get("grade"))        # None
    print(student.get("grade", "N/A"))  # N/A - قيمة افتراضية مخصصة
except (Exception, SystemExit) as _err:
    print(f"⚠️ {type(_err).__name__}: {_err}")
````

> ⚠️ **تحذير شائع**: المبتدئون يستخدمون `student["key"]` دائماً، فتنهار برامجهم فجأة بـ `KeyError` عند أول بيانات غير متوقعة. **القاعدة الذهبية**: إذا لم تكن متأكداً 100% أن المفتاح موجود، استخدم `.get()` دائماً.

````python
# التحقق من وجود مفتاح قبل استخدامه
if "grade" in student:
    print(student["grade"])
else:
    print("لا توجد درجة مسجلة")

# in تبحث في المفاتيح فقط، وليس القيم!
print("name" in student)   # True
print("Ahmed" in student)  # False (هذا بحث في القيم، وهو غير مدعوم مباشرة بـ in)
````

### إضافة وتعديل وحذف العناصر

````python
student = {"name": "Ahmed", "age": 22}

# الإضافة والتعديل بنفس الصيغة
student["age"] = 23           # تعديل (المفتاح موجود)
student["major"] = "AI"       # إضافة (المفتاح جديد)
print(student)
# {'name': 'Ahmed', 'age': 23, 'major': 'AI'}

# الحذف بطريقة del - تطلق خطأ إذا المفتاح غير موجود
del student["age"]

# الحذف بطريقة pop() - آمنة، وتُعيد القيمة المحذوفة
major = student.pop("major")
print(major)  # AI

# pop مع قيمة افتراضية لتجنب الخطأ
removed = student.pop("not_exist", "لا يوجد")
print(removed)  # لا يوجد

# حذف كل العناصر
student.clear()
print(student)  # {}
````

### التكرار على القواميس (Looping)

هذه من أهم المهارات العملية:

````python
scores = {"Sara": 90, "Omar": 75, "Lina": 88}

# التكرار على المفاتيح فقط (الافتراضي)
for name in scores:
    print(name)

# التكرار على القيم فقط
for score in scores.values():
    print(score)

# التكرار على المفاتيح والقيم معاً - الأكثر استخداماً عملياً
for name, score in scores.items():
    print(f"{name}: {score}")
````

> 🔄 **مقارنة**: في JavaScript تكتب `for (const [key, value] of Object.entries(obj))`، وفي Python الطريقة أبسط وأوضح: `for key, value in dict.items()`. هذا أحد الأسباب التي تجعل Python محبوبة في تعليم البرمجة.

### أهم Methods القاموس - جدول مرجعي شامل

| Method | الوصف | مثال |
|---|---|---|
| `.get(key, default)` | إرجاع القيمة أو قيمة افتراضية | `d.get("x", 0)` |
| `.keys()` | إرجاع كل المفاتيح | `d.keys()` |
| `.values()` | إرجاع كل القيم | `d.values()` |
| `.items()` | إرجاع أزواج (مفتاح, قيمة) | `d.items()` |
| `.pop(key, default)` | حذف وإرجاع القيمة | `d.pop("x")` |
| `.update(other_dict)` | دمج قاموس آخر (الكتابة فوق المشترك) | `d.update({"y": 2})` |
| `.setdefault(key, default)` | إرجاع القيمة، وإنشاؤها إن لم توجد | `d.setdefault("x", [])` |
| `.copy()` | نسخة سطحية | `d.copy()` |
| `.clear()` | حذف كل العناصر | `d.clear()` |

````python
# .update() - دمج قاموسين
defaults = {"theme": "dark", "lang": "ar", "font_size": 14}
user_settings = {"lang": "en"}

defaults.update(user_settings)
print(defaults)
# {'theme': 'dark', 'lang': 'en', 'font_size': 14}

# .setdefault() - مفيدة جداً لتجميع البيانات
word_groups: dict[str, list[str]] = {}
words = ["apple", "ant", "banana", "bee", "avocado"]

for word in words:
    first_letter = word[0]
    word_groups.setdefault(first_letter, []).append(word)

print(word_groups)
# {'a': ['apple', 'ant', 'avocado'], 'b': ['banana', 'bee']}
````

> 💡 **نصيحة الخبراء**: النمط أعلاه (`setdefault` لتجميع عناصر في قوائم فرعية) نمط شائع جداً في معالجة البيانات. هناك بديل أنظف وأسرع له اسم `defaultdict` من مكتبة `collections`:

````python
from collections import defaultdict

word_groups = defaultdict(list)
for word in words:
    word_groups[word[0]].append(word)  # لا حاجة لـ setdefault!

print(dict(word_groups))
````

### Dictionary Comprehension

تماماً مثل List Comprehension، لكن لإنشاء قواميس:

````python
# الصيغة: {key_expr: value_expr for item in iterable}

# مثال 1: تربيع الأرقام
numbers = [1, 2, 3, 4, 5]
squares = {n: n**2 for n in numbers}
print(squares)  # {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}

# مثال 2: مع شرط
even_squares = {n: n**2 for n in numbers if n % 2 == 0}
print(even_squares)  # {2: 4, 4: 16}

# مثال 3: عكس قاموس (قلب المفاتيح والقيم)
original = {"a": 1, "b": 2, "c": 3}
reversed_dict = {value: key for key, value in original.items()}
print(reversed_dict)  # {1: 'a', 2: 'b', 3: 'c'}

# مثال 4: تحويل قائمتين متوازيتين إلى قاموس باستخدام zip
names = ["Sara", "Omar", "Lina"]
scores = [90, 75, 88]
result = {name: score for name, score in zip(names, scores)}
print(result)  # {'Sara': 90, 'Omar': 75, 'Lina': 88}
````

> ⚠️ **خطأ شائع**: استخدام قيمة قابلة للتعديل (mutable) كمفتاح:

````python
# d = {[1, 2]: "value"}  # TypeError: unhashable type: 'list'
d = {(1, 2): "value"}  # صحيح! tuple غير قابلة للتعديل، لذا مسموحة كمفتاح
````

السبب التقني: المفاتيح يجب أن تكون **Hashable** (قابلة للتحويل إلى قيمة hash ثابتة)، والقيم القابلة للتعديل (list, dict, set) لا تملك hash ثابتاً لأن محتواها قد يتغير.

## 9.3 المجموعات (Sets)

### ما هي المجموعة؟

المجموعة هي تجميعة من عناصر **فريدة** (بدون تكرار) **وغير مرتبة**. فكّر بها كصندوق تضع فيه أشياء، لكن لا يمكن أن يحتوي على نفس الشيء مرتين، ولا يهمك ترتيبه.

````python
fruits: set[str] = {"apple", "banana", "apple", "orange"}
print(fruits)  
# {'apple', 'banana', 'orange'}  -- لاحظ التكرار اختفى تلقائياً!
````

هذه هي الاستخدام الأول والأشهر للمجموعات: **إزالة التكرار بسهولة**.

````python
numbers = [1, 2, 2, 3, 4, 4, 4, 5]
unique_numbers = list(set(numbers))
print(unique_numbers)  # [1, 2, 3, 4, 5] (الترتيب قد يختلف)
````

> ⚠️ **تحذير**: المجموعات **غير مرتبة**، فلا تعتمد على ترتيب معين عند تحويل قائمة إلى set ثم إعادتها إلى list. إذا احتجت الحفاظ على الترتيب مع إزالة التكرار، استخدم:

````python
numbers = [3, 1, 2, 3, 1, 4]
unique_ordered = list(dict.fromkeys(numbers))
print(unique_ordered)  # [3, 1, 2, 4] - الترتيب محفوظ!
````

### إنشاء المجموعات

````python
# الأقواس المعقوفة (بدون نقطتين - هذا ما يميزها عن dict)
s1: set[int] = {1, 2, 3}

# دالة set() - ضرورية للمجموعة الفارغة!
s2 = set()  

# ⚠️ تحذير حرج
empty = {}        # هذا قاموس فارغ وليس مجموعة فارغة!
print(type(empty))  # <class 'dict'>

# set() من قائمة
s3 = set([1, 2, 2, 3])
print(s3)  # {1, 2, 3}
````

> ⚠️ **خطأ شائع جداً عند المبتدئين**: كتابة `{}` معتقدين أنها مجموعة فارغة، بينما Python تفسرها كقاموس فارغ دائماً. **لإنشاء مجموعة فارغة استخدم `set()` فقط.**

### العمليات الأساسية على المجموعات

````python
colors: set[str] = {"red", "green", "blue"}

# إضافة عنصر واحد
colors.add("yellow")

# إضافة عدة عناصر
colors.update(["purple", "pink"])

# حذف عنصر - يطلق خطأ إذا غير موجود
colors.remove("red")

# حذف آمن - لا يطلق خطأ
colors.discard("not_exist")  # لا شيء يحدث

# حذف عنصر عشوائي وإرجاعه
item = colors.pop()

# التحقق من وجود عنصر - سريع جداً (انظر الأداء أدناه)
print("blue" in colors)
````

### العمليات الرياضية على المجموعات (القوة الحقيقية)

هنا يكمن السر الحقيقي لأهمية المجموعات: **عمليات نظرية المجموعات الرياضية**.

````python
python_devs = {"Ahmed", "Sara", "Omar", "Lina"}
js_devs = {"Omar", "Lina", "Khaled", "Noor"}

# الاتحاد (Union): كل العناصر من الاثنين بدون تكرار
all_devs = python_devs | js_devs
# أو: python_devs.union(js_devs)
print(all_devs)
# {'Ahmed', 'Sara', 'Omar', 'Lina', 'Khaled', 'Noor'}

# التقاطع (Intersection): العناصر المشتركة فقط
both_devs = python_devs & js_devs
# أو: python_devs.intersection(js_devs)
print(both_devs)
# {'Omar', 'Lina'}

# الفرق (Difference): في الأولى وليس في الثانية
only_python = python_devs - js_devs
# أو: python_devs.difference(js_devs)
print(only_python)
# {'Ahmed', 'Sara'}

# الفرق التماثلي (Symmetric Difference): في أحدهما فقط وليس الاثنين
exclusive = python_devs ^ js_devs
# أو: python_devs.symmetric_difference(js_devs)
print(exclusive)
# {'Ahmed', 'Sara', 'Khaled', 'Noor'}
````

### لماذا المجموعات سريعة جداً؟

> 🔄 **مقارنة الأداء - نظرة تحت غطاء المحرك**

| العملية | List | Set |
|---|---|---|
| البحث عن عنصر (`x in collection`) | O(n) - يفحص كل عنصر تباعاً | O(1) - وصول مباشر تقريباً |
| إضافة عنصر | O(1) في الآخر | O(1) |
| إزالة التكرار | تحتاج كود يدوي | تلقائي بالتعريف |

````python
import time

big_list = list(range(1_000_000))
big_set = set(range(1_000_000))

start = time.perf_counter()
999_999 in big_list
print(f"List: {time.perf_counter() - start:.6f} ثانية")

start = time.perf_counter()
999_999 in big_set
print(f"Set: {time.perf_counter() - start:.6f} ثانية")

# النتيجة التقريبية: List أبطأ بآلاف المرات!
````

السبب التقني: القوائم تبحث عنصراً تلو الآخر (Linear Search)، بينما المجموعات والقواميس تستخدم بنية بيانات تسمى **Hash Table** تحسب «عنوان» كل عنصر مباشرة، فتصل إليه فوراً تقريباً بغض النظر عن حجم المجموعة.

> 💡 **نصيحة الخبراء**: إذا كان كودك يحتوي على سطر مثل `if x in my_list:` داخل حلقة تتكرر آلاف المرات، وكانت `my_list` كبيرة، **حوّلها إلى set أولاً**:

```python
# بطيء مع قوائم كبيرة
allowed_ids = [1, 5, 10, 15, ..., 50000]  # قائمة ضخمة
for user_id in incoming_ids:
    if user_id in allowed_ids:  # بحث خطي بطيء كل مرة!
        process(user_id)

# سريع جداً
allowed_ids_set = set(allowed_ids)  # تحويل مرة واحدة
for user_id in incoming_ids:
    if user_id in allowed_ids_set:  # بحث فوري!
        process(user_id)
```

🧾 **للقراءة فقط** — مثال مقتطف يعتمد على `incoming_ids`، `process` من مثال سابق.

## 9.4 جدول المقارنة الشامل: متى تستخدم ماذا؟

هذا هو الجدول الذي وعدناك به في نهاية الفصل السابق. ادرسه جيداً، وارجع إليه كلما ترددت في اختيار هيكل البيانات المناسب.

| الخاصية | **List** `[]` | **Tuple** `()` | **Dict** `{k: v}` | **Set** `{}` |
|---|---|---|---|---|
| **قابل للتعديل؟** | ✅ نعم | ❌ لا | ✅ نعم | ✅ نعم |
| **مرتب (يحفظ الترتيب)؟** | ✅ نعم | ✅ نعم | ✅ نعم (منذ 3.7) | ❌ لا |
| **يسمح بالتكرار؟** | ✅ نعم | ✅ نعم | المفاتيح: ❌ / القيم: ✅ | ❌ لا |
| **الوصول للعنصر** | بالموقع `[0]` | بالموقع `[0]` | بالمفتاح `["key"]` | لا يوجد وصول مباشر |
| **سرعة البحث `in`** | بطيء O(n) | بطيء O(n) | سريع جداً O(1) | سريع جداً O(1) |
| **الاستخدام المثالي** | تسلسل عناصر قد يتغير (قائمة مهام، نتائج) | بيانات ثابتة لا تتغير (إحداثيات، تاريخ ميلاد) | بيانات مرتبطة بأسماء/معرّفات (سجل مستخدم، إعدادات) | ضمان التفرد، عمليات رياضية بين مجموعات |
| **مثال واقعي** | `cart_items = ["pen", "book"]` | `point = (3, 4)` | `user = {"name": "Ali"}` | `tags = {"python", "web"}` |
| **Comprehension** | `[x for x in ...]` | غير مباشر (generator) | `{k: v for ...}` | `{x for x in ...}` |

### 🧭 خريطة قرار سريعة

```
هل البيانات ثابتة لن تتغير أبداً؟
├── نعم → Tuple
└── لا
    ├── هل أحتاج الوصول بالاسم/المفتاح وليس بالرقم؟
    │   └── نعم → Dict
    └── لا
        ├── هل يهمني منع التكرار أو عمل عمليات رياضية (تقاطع/اتحاد)؟
        │   └── نعم → Set
        └── لا → List (الخيار الافتراضي الأكثر مرونة)
```

> 🔄 **مقارنة مع لغات أخرى**: في JavaScript، `Object` يقابل `dict`، و`Array` يقابل `list`، و`Set` موجود بنفس الاسم والمفهوم تقريباً. في Java، `HashMap` يقابل `dict`، و`HashSet` يقابل `set`. المفاهيم عالمية؛ فقط الأسماء والصياغة تختلف.

## 9.5 مشروع تطبيقي: نظام إدارة مخزون بسيط

سنبني نظاماً لإدارة مخزون متجر صغير، يستخدم القواميس للمنتجات، والمجموعات لتتبع الفئات الفريدة.

**المتطلبات:**
1. تخزين المنتجات: الاسم، السعر، الكمية، الفئة
2. إضافة منتج جديد
3. تحديث كمية منتج (بيع/شراء)
4. عرض كل المنتجات التي كميتها أقل من حد معين (تنبيه نفاد المخزون)
5. عرض كل الفئات الفريدة الموجودة في المخزون
6. حساب القيمة الإجمالية للمخزون

```python
inventory: dict[str, dict[str, object]] = {}

def add_product(name: str, price: float, quantity: int, category: str) -> None:
    """إضافة منتج جديد إلى المخزون."""
    if name in inventory:
        print(f"⚠️  المنتج '{name}' موجود مسبقاً. استخدم update_quantity للتعديل.")
        return
    inventory[name] = {
        "price": price,
        "quantity": quantity,
        "category": category
    }
    print(f"✅ تمت إضافة '{name}' بنجاح.")

def update_quantity(name: str, change: int) -> None:
    """تحديث كمية منتج (موجب للشراء، سالب للبيع)."""
    if name not in inventory:
        print(f"❌ المنتج '{name}' غير موجود.")
        return

    new_quantity = inventory[name]["quantity"] + change
    if new_quantity < 0:
        print(f"❌ لا توجد كمية كافية من '{name}'.")
        return

    inventory[name]["quantity"] = new_quantity
    print(f"✅ كمية '{name}' الآن: {new_quantity}")

def low_stock_alert(threshold: int = 5) -> list[str]:
    """إرجاع قائمة بأسماء المنتجات التي كميتها أقل من الحد المسموح."""
    low_items = [
        name for name, info in inventory.items()
        if info["quantity"] < threshold
    ]
    return low_items

def unique_categories() -> set[str]:
    """إرجاع كل الفئات الفريدة الموجودة في المخزون."""
    return {info["category"] for info in inventory.values()}

def total_inventory_value() -> float:
    """حساب القيمة الإجمالية لكل المخزون (السعر × الكمية)."""
    total = sum(
        info["price"] * info["quantity"]
        for info in inventory.values()
    )
    return total

def print_report() -> None:
    """طباعة تقرير شامل عن حالة المخزون."""
    print("\n" + "=" * 40)
    print("📊 تقرير المخزون")
    print("=" * 40)

    for name, info in inventory.items():
        print(f"- {name}: {info['quantity']} قطعة | "
              f"{info['price']:.2f} ر.س | الفئة: {info['category']}")

    print("-" * 40)
    print(f"💰 القيمة الإجمالية: {total_inventory_value():,.2f} ر.س")
    print(f"🏷️  الفئات المتاحة: {', '.join(unique_categories())}")

    low_items = low_stock_alert(threshold=5)
    if low_items:
        print(f"⚠️  تنبيه نفاد المخزون: {', '.join(low_items)}")
    else:
        print("✅ لا توجد منتجات على وشك النفاد.")
    print("=" * 40)

# --- تشغيل تجريبي للنظام ---
if __name__ == "__main__":
    add_product("لابتوب", 3500.0, 10, "إلكترونيات")
    add_product("ماوس", 50.0, 3, "إلكترونيات")
    add_product("كرسي مكتب", 450.0, 8, "أثاث")
    add_product("طاولة", 700.0, 2, "أثاث")

    update_quantity("لابتوب", -2)   # بيع لابتوبين
    update_quantity("ماوس", 10)     # شراء 10 ماوسات جديدة

    print_report()
```

````python
def compare_inventories(inv_a: dict, inv_b: dict) -> dict[str, set]:
    """مقارنة منتجات مخزونين مختلفين (مثلاً فرعين لنفس المتجر)."""
    products_a = set(inv_a.keys())
    products_b = set(inv_b.keys())

    return {
        "only_in_a": products_a - products_b,
        "only_in_b": products_b - products_a,
        "in_both": products_a & products_b
    }

branch_1 = {"لابتوب": {}, "ماوس": {}, "كرسي": {}}
branch_2 = {"ماوس": {}, "كرسي": {}, "طابعة": {}}

result = compare_inventories(branch_1, branch_2)
print(result)
# {'only_in_a': {'لابتوب'}, 'only_in_b': {'طابعة'}, 'in_both': {'ماوس', 'كرسي'}}
````

اطبع: من في الفريقين معاً، من في A فقط، من في B فقط، ومن في فريق واحد فقط (وليس الاثنين).

**تمرين 6**: اكتب دالة `merge_dicts(d1: dict, d2: dict) -> dict` تدمج قاموسين، وفي حال وجود مفتاح مشترك، تجمع القيم (بافتراض أنها أرقام).

<details>
<summary>✅ الحل</summary>

````python
# تمرين 4
def count_words(text: str) -> dict[str, int]:
    words = text.lower().split()
    counts: dict[str, int] = {}
    for word in words:
        counts[word] = counts.get(word, 0) + 1
    return counts

result = count_words("Python is great and Python is fun")
print(result)
# {'python': 2, 'is': 2, 'great': 1, 'and': 1, 'fun': 1}

# تمرين 5
team_a = {"Ali", "Sara", "Omar"}
team_b = {"Omar", "Lina", "Khaled"}

print("في الفريقين:", team_a & team_b)
print("في A فقط:", team_a - team_b)
print("في B فقط:", team_b - team_a)
print("في فريق واحد فقط:", team_a ^ team_b)

# تمرين 6
def merge_dicts(d1: dict, d2: dict) -> dict:
    result = d1.copy()
    for key, value in d2.items():
        result[key] = result.get(key, 0) + value
    return result

a = {"apples": 5, "bananas": 3}
b = {"bananas": 2, "oranges": 7}
print(merge_dicts(a, b))
# {'apples': 5, 'bananas': 5, 'oranges': 7}
````

### 🔴 تحدي

**تمرين 7**: اكتب برنامجاً يقرأ قائمة طلبات (كل طلب عبارة عن قاموس فيه `product` و`quantity` و`price`)، ويحسب:
- إجمالي المبيعات لكل منتج
- المنتج الأكثر مبيعاً (بالكمية)
- استخدم `defaultdict` من `collections` في الحل.

<details>
<summary>✅ الحل</summary>

````python
from collections import defaultdict

orders = [
    {"product": "قلم", "quantity": 10, "price": 2.0},
    {"product": "دفتر", "quantity": 5, "price": 8.0},
    {"product": "قلم", "quantity": 7, "price": 2.0},
    {"product": "ممحاة", "quantity": 20, "price": 1.0},
    {"product": "دفتر", "quantity": 3, "price": 8.0},
]

sales_total: dict[str, float] = defaultdict(float)
quantity_total: dict[str, int] = defaultdict(int)

for order in orders:
    product = order["product"]
    sales_total[product] += order["quantity"] * order["price"]
    quantity_total[product] += order["quantity"]

print("💰 إجمالي المبيعات لكل منتج:")
for product, total in sales_total.items():
    print(f"  {product}: {total:.2f} ر.س")

best_seller = max(quantity_total, key=quantity_total.get)
print(f"\n🏆 المنتج الأكثر مبيعاً: {best_seller} ({quantity_total[best_seller]} قطعة)")
````

## 9.7 اختبر فهمك (Quiz)

<details>
<summary>❓ السؤال 1: ما الفرق بين `{}` و `set()`؟</summary>

**الإجابة**: `{}` تُنشئ **قاموساً فارغاً** دائماً في Python، بينما `set()` هي الطريقة الوحيدة لإنشاء **مجموعة فارغة**. هذا من أكثر الأخطاء شيوعاً عند المبتدئين.
</details>

<details>
<summary>❓ السؤال 2: لماذا لا يمكن استخدام list كمفتاح في dict؟</summary>

**الإجابة**: لأن المفاتيح يجب أن تكون **Hashable** (قابلة لحساب قيمة hash ثابتة لا تتغير). القوائم قابلة للتعديل (mutable)، فلو سمحت Python باستخدامها كمفتاح، ثم عدّلتها، سيختل موقعها في بنية الـ Hash Table الداخلية. لهذا فقط الأنواع غير القابلة للتعديل (str, int, float, tuple, frozenset) مسموحة كمفاتيح.
</details>

<details>
<summary>❓ السؤال 3: ما ناتج الكود التالي؟</summary>

````python
a = {1, 2, 3}
b = {3, 4, 5}
print(a ^ b)
````

**الإجابة**: `{1, 2, 4, 5}` — هذا هو الفرق التماثلي (Symmetric Difference): كل العناصر الموجودة في أحد المجموعتين فقط، وليس في كليهما.
</details>

<details>
<summary>❓ السؤال 4: متى تفضل استخدام .get() بدلاً من []؟</summary>

**الإجابة**: عندما لا تكون متأكداً 100% من وجود المفتاح في القاموس. `[]` تطلق `KeyError` إذا لم يوجد المفتاح وتُوقف البرنامج، بينما `.get()` تُرجع `None` (أو قيمة افتراضية تحددها أنت) بأمان دون توقف البرنامج.
</details>

<details>
<summary>❓ السؤال 5: ما الفرق بين البحث عن عنصر في list مقابل set من ناحية الأداء؟</summary>

**الإجابة**: البحث في list هو O(n) — أي أن Python تفحص العناصر واحداً تلو الآخر حتى تجد المطلوب أو تصل للنهاية، فكلما كبرت القائمة طال وقت البحث. البحث في set (وكذلك dict) هو O(1) تقريباً — أي وقت شبه ثابت بغض النظر عن حجم المجموعة، بفضل بنية Hash Table الداخلية.
</details>
