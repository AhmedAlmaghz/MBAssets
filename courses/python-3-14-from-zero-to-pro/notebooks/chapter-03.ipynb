## 3.1 📖 مقدمة: لماذا المتغيرات هي نقطة البداية الحقيقية؟

في الفصل السابق، كتبت برنامجك الأول الذي يسأل عن اسمك ويطبع ترحيبًا. لكن هل تساءلت: كيف «تذكّر» البرنامج اسمك بين سطر `input()` وسطر `print()`؟ الإجابة: **المتغيرات**.

المتغيرات هي أبسط وأهم مفهوم في أي لغة برمجة على الإطلاق. بدونها، لا يمكن للبرنامج تخزين أي معلومة، ولا معالجتها، ولا استخدامها لاحقًا. كل شيء ستتعلمه لاحقًا - من القوائم إلى الدوال إلى البرمجة الكائنية - مبني فوق هذا المفهوم الأساسي.

## 3.2 ما هو المتغير؟ (الفهم الصحيح، لا المبسّط الخاطئ)

### 🎁 التشبيه الشائع (ولماذا هو غير دقيق تمامًا في Python)

أغلب الكتب تشرح المتغير كـ«صندوق» تضع بداخله قيمة:

```
[صندوق باسم "age"] ← يحتوي 25
```

هذا التشبيه مفيد للبداية، لكنه **غير دقيق** بخصوص كيفية عمل Python فعليًا، وسيُسبب لك التباسًا لاحقًا عند تعلّم القوائم والكائنات. لذلك، سنتعلّم النموذج الصحيح من البداية.

### 🏷️ النموذج الصحيح: المتغيرات كـ«ملصقات» (Labels)

في Python، الصورة الأدق هي:

> **المتغير ليس صندوقًا يحتوي قيمة، بل «ملصق» (اسم) يُشير إلى مكان في الذاكرة تُخزَّن فيه القيمة.**

تخيّل الأمر هكذا:

````python
age = 25
````

ما يحدث فعليًا في الذاكرة:

```
الذاكرة:
┌─────────────┐
│   القيمة 25  │  ← كائن (Object) مُخزَّن في مكان ما بالذاكرة
└─────────────┘
       ↑
       │ (يُشير إليه)
       │
   الملصق "age"
```

الاسم `age` هو مجرد **ملصق** يُشير إلى حيث يوجد الرقم 25 في الذاكرة. هذا الفرق يبدو بسيطًا الآن، لكنه سيكون حاسمًا لفهم سلوكيات معينة لاحقًا (خصوصًا عند التعامل مع القوائم في الفصل السابع).

### 🔬 إثبات عملي: دالة `id()`

يمكنك فعليًا «رؤية» أين تُخزَّن القيمة في الذاكرة باستخدام دالة `id()` المُدمجة:

````python
age = 25
print(id(age))  # سيطبع رقمًا طويلاً مثل: 140712834567280
````

هذا الرقم هو عنوان الذاكرة (مبسّط) حيث يوجد الكائن `25`. جرّب هذا:

````python
age = 25
age_copy = age

print(id(age))       # 140712834567280
print(id(age_copy))  # 140712834567280  ← نفس العنوان بالضبط!
````

> 💡 **ملاحظة مهمة**: كلا المتغيرين `age` و `age_copy` يُشيران لنفس الكائن `25` في الذاكرة. لم يتم «نسخ» الرقم فعليًا، بل تم إنشاء ملصق جديد يُشير لنفس المكان. هذا السلوك طبيعي تمامًا للأرقام الصغيرة والقيم البسيطة، وسنتعمّق في تبعاته لاحقًا مع أنواع بيانات أكثر تعقيدًا.

## 3.3 إنشاء المتغيرات: عملية الإسناد (Assignment)

### ✍️ الصيغة الأساسية

```python
variable_name = value
```

🧾 **للقراءة فقط** — قالب يوضّح الصيغة العامة، وليس كودًا جاهزًا.

علامة `=` هنا تُسمى **عامل الإسناد (Assignment Operator)**، وهي **لا تعني** «يساوي» بالمعنى الرياضي إطلاقًا، بل تعني: **«اجعل هذا الاسم يُشير إلى تلك القيمة»**.

````python
name = "سارة"
age = 30
height = 1.65
is_student = False
````

### ⚠️ تحذير حرج: الخلط بين `=` و `==`

هذا من أكثر الأخطاء شيوعًا بين المبتدئين على الإطلاق، ويستمر معهم أحيانًا لأسابيع قبل أن يستوعبوه تمامًا:

| الرمز | الاسم | الوظيفة | مثال |
|---|---|---|---|
| `=` | Assignment (إسناد) | **تخزين** قيمة في متغير | `age = 25` |
| `==` | Equality (مقارنة) | **مقارنة** إن كانت قيمتان متساويتين | `age == 25` |

```python
age = 25        # ✅ صحيح: تخزين القيمة 25 في age
print(age == 25)  # ✅ صحيح: مقارنة - هل age تساوي 25؟ → True

age = 25
print(age = 25)  # ❌ خطأ! SyntaxError - لا يمكن استخدام = داخل print()
```

⚠️ **للقراءة فقط** — خطأ مقصود للتوضيح، لذا لا يُنفَّذ.

> 🎯 **قاعدة ذهبية للتذكّر**: استخدم `=` عندما تريد «**تخزين**» قيمة، واستخدم `==` عندما تريد «**السؤال**» إن كانت قيمتان متساويتين. سنتعمّق في `==` أكثر في فصل التحكم بسير البرنامج (الفصل السادس).

## 3.4 قواعد وأعراف تسمية المتغيرات

### 📏 القواعد الإلزامية (يجب اتباعها، وإلا سيحدث خطأ)

Python تفرض قواعد صارمة لأسماء المتغيرات:

```python
# ✅ أسماء صحيحة
name = "أحمد"
age2 = 25
_private_value = 100
userName = "test"

# ❌ أسماء غير صحيحة (ستُسبب SyntaxError)
2age = 25        # لا يمكن البدء برقم
user-name = "x"  # الشرطة - غير مسموحة
class = "test"   # "class" كلمة محجوزة في Python
```

⚠️ **للقراءة فقط** — خطأ مقصود للتوضيح، لذا لا يُنفَّذ.

**القواعد الإلزامية:**
1. يجب أن يبدأ الاسم بحرف أو شرطة سفلية `_` (لا يبدأ برقم أبدًا)
2. يمكن أن يحتوي على حروف، أرقام، وشرطات سفلية فقط (لا مسافات، لا رموز خاصة)
3. حساس لحالة الأحرف (`Name` و `name` متغيران مختلفان تمامًا)
4. لا يمكن استخدام **الكلمات المحجوزة** (Reserved Keywords) مثل `if`, `for`, `class`, `True`, `None`...

### 🎨 الأعراف الاحترافية (Convention) - ليست إلزامية، لكن اتبعها دائمًا

هنا يأتي الفرق بين كود «يعمل» وكود «احترافي». المجتمع العالمي لمبرمجي Python يتبع معيارًا موثقًا رسميًا يُسمى **PEP 8**، وأهم قاعدة فيه لتسمية المتغيرات:

> **استخدم snake_case دائمًا**: كل الحروف صغيرة، والكلمات مفصولة بشرطة سفلية.

```python
# ✅ الأسلوب الاحترافي الموصى به (snake_case)
first_name = "محمد"
total_price = 150.5
is_logged_in = True

# ❌ أساليب تعمل لكنها غير احترافية في Python
firstName = "محمد"      # camelCase - شائع في JavaScript وليس Python
FirstName = "محمد"      # PascalCase - يُستخدم للـ Classes فقط لاحقًا
FIRST_NAME = "محمد"     # كله كبير - يُستخدم للثوابت (Constants) فقط
```

⚠️ **للقراءة فقط** — خطأ مقصود للتوضيح، لذا لا يُنفَّذ.

### 🔄 مقارنة سريعة مع لغات أخرى

| اللغة | العُرف الشائع | مثال |
|---|---|---|
| **Python** | snake_case | `user_name` |
| **JavaScript** | camelCase | `userName` |
| **Java** | camelCase | `userName` |

> ⚠️ **تحذير للمبرمجين القادمين من لغات أخرى**: إذا كنت تعرف JavaScript أو Java مسبقًا، ستميل عفويًا لكتابة `camelCase`. تجنّب هذا في Python - الكود الذي لا يتبع snake_case يُعتبر «كودًا غير أصيل» (not Pythonic) في أي مراجعة كود احترافية.

### 📛 الأسماء ذات المعنى: فرق بين مبرمج مبتدئ ومحترف

```python
# ❌ أسماء غامضة (تعمل لكنها سيئة)
x = 25
y = "محمد"
z = x * 12

# ✅ أسماء واضحة ومعبّرة (احترافية)
age = 25
name = "محمد"
age_in_months = age * 12
```

⚠️ **للقراءة فقط** — خطأ مقصود للتوضيح، لذا لا يُنفَّذ.

> 💡 **نصيحة الخبراء**: الكود يُقرأ أكثر بكثير مما يُكتب. قد تعود لكودك بعد 6 أشهر ولن تتذكّر ماذا كان يعني `x`. استثمر ثانيتين إضافيتين الآن في اختيار اسم واضح، ستوفر عليك دقائق (أو ساعات) من الحيرة لاحقًا.

## 3.5 الأنواع الأساسية للبيانات (Data Types)

Python تحتوي أنواعًا كثيرة من البيانات، لكن في هذا الفصل سنتعرّف على **الأنواع الأساسية (Primitive Types)** التي ستستخدمها يوميًا.

### 🔢 الأعداد الصحيحة (Integer - `int`)

أي رقم صحيح، موجب أو سالب، بدون فاصلة عشرية:

````python
age = 25
temperature = -5
year = 2025

print(type(age))  # <class 'int'>
````

> 💡 **ميزة فريدة في Python**: على عكس لغات أخرى كثيرة، أعداد `int` في Python ليس لها حد أقصى للحجم (محدودة فقط بذاكرة جهازك). يمكنك فعليًا تخزين أرقام بمئات الخانات دون أي مشاكل:

````python
huge_number = 123456789012345678901234567890
print(huge_number * 2)  # يعمل بدون أي مشاكل!
````

### 🔢 الأعداد العشرية (Float)

أي رقم يحتوي فاصلة عشرية:

````python
height = 1.75
price = 99.99
pi_approx = 3.14159

print(type(height))  # <class 'float'>
````

> ⚠️ **تحذير مهم - فخ الدقة العشرية**: ستُفاجأ لاحقًا (في الفصل الخامس) بأن Python (وكل لغات البرمجة تقريبًا) تعاني من مشكلة دقة في الأعداد العشرية:

````python
print(0.1 + 0.2)  # 0.30000000000000004  ← ليس 0.3 تمامًا!
````

هذا **ليس خطأ** في Python، بل طبيعة كيفية تخزين الكمبيوترات للأعداد العشرية في الذاكرة (نظام ثنائي). سنتعلّم كيف نتعامل مع هذا بذكاء في الفصل الخامس.

### 📝 النصوص (String - `str`)

أي سلسلة من الحروف، محاطة بعلامات اقتباس:

````python
name = "سارة"
city = 'الرياض'  # علامات مفردة أو مزدوجة - كلاهما صحيح
message = "اليوم هو يوم جميل!"

print(type(name))  # <class 'str'>
````

> 💡 **نصيحة اتساق**: Python لا تفرّق بين `'single quotes'` و `"double quotes"` وظيفيًا، لكن اختر أسلوبًا واحدًا والتزم به في كل مشروعك للاتساق. المعيار الأكثر شيوعًا عالميًا هو علامات الاقتباس المزدوجة `"..."`.

### ✅ القيم المنطقية (Boolean - `bool`)

قيمتان فقط: `True` أو `False` (لاحظ حرف كبير إلزامي في البداية):

````python
is_student = True
is_raining = False

print(type(is_student))  # <class 'bool'>
````

> ⚠️ **خطأ شائع جدًا**: كتابة `true` أو `false` بحروف صغيرة (كما في JavaScript أو Java) ستُسبب `NameError` في Python:

```python
is_active = true  # ❌ NameError: name 'true' is not defined

```

⚠️ **للقراءة فقط** — خطأ مقصود للتوضيح، لذا لا يُنفَّذ.

### 🚫 القيمة الفارغة (`None`)

تمثّل «لا شيء» أو «غياب القيمة» - مفهوم مهم جدًا سنستخدمه كثيرًا لاحقًا:

````python
middle_name = None  # الشخص ليس لديه اسم أوسط

print(type(middle_name))  # <class 'NoneType'>
````

> 💡 **متى تُستخدم `None`؟** عندما تريد التعبير عن «لا توجد قيمة بعد» أو «فارغ عمدًا»، على عكس الصفر `0` أو النص الفارغ `""` اللذين هما قيم فعلية موجودة. سنرى استخدامات عملية قوية لهذا في فصول الدوال والبرمجة الكائنية.

## 3.6 Python لغة ديناميكية التنميط (Dynamically Typed)

### 🔄 الفهم الجوهري

في لغات مثل C++ أو Java، يجب عليك **تحديد نوع المتغير صراحة** عند إنشائه، ولا يمكن تغييره لاحقًا:

```java
// في Java (ليست Python!)
int age = 25;
age = "خمسة وعشرون";  // ❌ خطأ! لا يمكن تغيير النوع
```

🧾 **للقراءة فقط** — محتوى ملف `java`.

لكن في Python، النوع **يُحدَّد تلقائيًا** حسب القيمة، ويمكن أن **يتغيّر** في أي وقت:

````python
age = 25         # الآن age من نوع int
print(type(age))  # <class 'int'>

age = "خمسة وعشرون"  # ✅ مسموح تمامًا في Python!
print(type(age))      # <class 'str'>
````

### 🔄 مقارنة: ثابت التنميط مقابل ديناميكي التنميط

| الخاصية | Statically Typed (Java, C++) | Dynamically Typed (Python) |
|---|---|---|
| **تحديد النوع** | إلزامي وصريح عند الإنشاء | تلقائي حسب القيمة |
| **تغيير النوع لاحقًا** | ❌ غير مسموح | ✅ مسموح |
| **اكتشاف أخطاء الأنواع** | قبل التشغيل (compile time) | أثناء التشغيل فقط |
| **سرعة الكتابة** | أبطأ (تفاصيل أكثر) | ⚡ أسرع وأكثر مرونة |
| **احتمالية أخطاء مخفية** | أقل | ⚠️ أعلى نسبيًا |

> ⚠️ **تحذير احترافي مهم**: مرونة Python هذه سلاح ذو حدين. في مشاريع صغيرة، هي ميزة رائعة تُسرّع التطوير. لكن في مشاريع كبيرة بفرق متعددة، قد يُسبب هذا أخطاءً مخفية يصعب اكتشافها (مثل تمرير نص لدالة تتوقع رقمًا دون أي تحذير فوري). هذا بالضبط سبب ابتكار **Type Hints** التي سنتعلّمها في القسم القادم - وهي حل Python الأنيق لهذه المشكلة.

## 3.7 Type Hints: الكتابة الواضحة من اليوم الأول

### 🤔 لماذا نتعلّم هذا بهذا التبكير؟

معظم الكتب التعليمية تؤجّل هذا المفهوم لفصول متقدمة جدًا، أو تتجاهله كليًا. لكننا سنكسر هذا التقليد، لأن **Type Hints** أصبحت جزءًا أساسيًا من أسلوب كتابة Python الحديث والاحترافي، ومن الأفضل بناء العادة الصحيحة من البداية بدل تعلّم «الطريقة القديمة» ثم «تصحيحها» لاحقًا.

### ✍️ الصيغة الأساسية

```python
variable_name: type = value
```

🧾 **للقراءة فقط** — قالب يوضّح الصيغة العامة، وليس كودًا جاهزًا.

مقارنة بين الأسلوبين:

````python
# بدون Type Hint (يعمل تمامًا، لكن أقل وضوحًا)
age = 25
name = "محمد"

# مع Type Hint (نفس الوظيفة، لكن أوضح للقارئ والأدوات)
age: int = 25
name: str = "محمد"
is_active: bool = True
height: float = 1.75
````

### 🎯 ما الفائدة الفعلية؟

Type Hints **لا تُغيّر** سلوك الكود إطلاقًا وقت التشغيل (Python ستتجاهلها عمليًا)، لكنها توفر فوائد ضخمة:

1. **توثيق ذاتي للكود**: أي شخص (حتى أنت بعد 6 أشهر) يفهم نوع البيانات المتوقع فورًا
2. **اكتشاف أخطاء مبكرًا**: محررات مثل VS Code (مع Pylance) تُنبّهك فورًا إن حاولت استخدام نوع خاطئ
3. **إكمال تلقائي أذكى**: المحرر يقترح عليك الدوال والخصائص الصحيحة بناءً على النوع المُعلن

### 🔬 تجربة عملية في VS Code

````python
age: int = 25
age = "خمسة وعشرون"  # لن يُسبب خطأ وقت التشغيل، لكن VS Code سيُظهر خط تحذير أحمر تحته!
````

> 💡 **نصيحة عملية**: من الآن فصاعدًا في هذا الكتاب، سنستخدم Type Hints في كل مثال تقريبًا - ليس لأنها إلزامية، بل لأنها **أفضل ممارسة عالمية حديثة** ستجعل كودك أوضح واحترافيًا من البداية، وستجنّبك إعادة تعلّم عادات لاحقًا.

> ⚠️ **توضيح مهم**: Type Hints **اختيارية بالكامل** في Python (على عكس لغات مثل Java حيث هي إلزامية). يمكنك كتابة برنامج كامل بدونها وسيعمل تمامًا. لكن في عالم العمل الاحترافي اليوم (خصوصًا في الشركات الكبرى)، أصبحت شبه معيار إلزامي لأي كود جاد.

## 3.8 تحويل الأنواع (Type Casting)

أحيانًا تحتاج لتحويل قيمة من نوع لآخر صراحة. Python توفر دوال مُدمجة لهذا:

````python
# تحويل من str إلى int
age_text: str = "25"
age_number: int = int(age_text)
print(age_number + 5)  # 30

# تحويل من int إلى str
score: int = 95
score_text: str = str(score)
print("درجتك: " + score_text)  # درجتك: 95

# تحويل من str إلى float
price_text: str = "19.99"
price: float = float(price_text)

# تحويل إلى bool
print(bool(0))      # False
print(bool(1))      # True
print(bool(""))     # False (نص فارغ = False)
print(bool("لا"))   # True  (أي نص غير فارغ = True، حتى لو كان "لا"!)
````

### ⚠️ الفخ الأشهر على الإطلاق للمبتدئين: `input()` تُعيد دائمًا نصًا

تذكر برنامجك في الفصل الثاني؟ هذا الفخ بالضبط وقعت فيه (أو كاد):

```python
age = input("كم عمرك؟ ")  # المستخدم يكتب: 25
print(type(age))            # <class 'str'> ← ليس int!

# محاولة استخدامه كرقم مباشرة:
age_next_year = age + 1  # ❌ TypeError: can only concatenate str (not "int") to str
```

⚠️ **للقراءة فقط** — خطأ مقصود للتوضيح، لذا لا يُنفَّذ.

**الحل الصحيح:**

```python
age: str = input("كم عمرك؟ ")
age_number: int = int(age)  # التحويل الصريح إلى int
age_next_year: int = age_number + 1
print(age_next_year)  # يعمل بنجاح الآن
```

🧾 **للقراءة فقط** — يحتاج إدخالًا يدويًا.

> 🎯 **قاعدة ذهبية يجب حفظها**: دالة `input()` **تُعيد دائمًا نصًا (str)**، بغض النظر عمّا يكتبه المستخدم - حتى لو كتب رقمًا! يجب عليك دائمًا تحويله صراحة (`int()` أو `float()`) إن أردت استخدامه كرقم في عمليات حسابية.

### ⚠️ تحذير: محاولة تحويل نص غير رقمي


سنتعلّم في الفصل الحادي عشر (معالجة الأخطاء) كيف «نمسك» هذا النوع من الأخطاء بأناقة قبل أن يُوقف البرنامج بالكامل.

## 3.9 الإسناد المتعدد وتبديل القيم (تقنيات Python الأنيقة)

### 🎁 الإسناد المتعدد

Python تسمح بإسناد عدة متغيرات في سطر واحد بأناقة:

````python
# الطريقة التقليدية
x = 5
y = 10
z = 15

# طريقة Python الأنيقة (Multiple Assignment)
x, y, z = 5, 10, 15
````

### 🎁 إسناد نفس القيمة لعدة متغيرات

````python
a = b = c = 0
print(a, b, c)  # 0 0 0
````

### 🔄 تبديل قيم متغيرين (دون متغير مؤقت!)

````python
# الطريقة التقليدية (في لغات أخرى، أو حتى في Python)
temp = x
x = y
y = temp
````

## 3.10 الثوابت (Constants): عُرف لا قاعدة صارمة

### 🤔 هل توجد ثوابت حقيقية في Python؟

بصراحة تامة: **لا**. على عكس لغات مثل Java أو C++ التي لديها كلمة `const` تمنع تغيير القيمة فعليًا، Python **لا تملك آلية حقيقية** لمنع تغيير أي متغير.

لكن المجتمع البرمجي اتفق على **عُرف (Convention)**: أي متغير بأحرف كبيرة بالكامل (`UPPER_CASE`) يُعتبر «ثابتًا» **بالاتفاق**، ولا يجب تغييره، رغم أن Python تقنيًا لن تمنعك:

````python
PI: float = 3.14159
MAX_USERS: int = 100
APP_NAME: str = "نظام إدارة المكتبة"

# تقنيًا يمكنك فعل هذا، لكنه يُعتبر خطأ في الممارسة الاحترافية:
PI = 3.14  # ⚠️ ليس خطأ تقنيًا، لكنه مخالفة صارخة للعُرف المتفق عليه
````

> ✅ **أفضل الممارسات**: استخدم `UPPER_CASE` لأي قيمة تعرف أنها لن (ويجب ألا) تتغيّر طوال تشغيل البرنامج (مثل Pi، الحد الأقصى لشيء ما، اسم التطبيق). هذا يُرسل إشارة واضحة لأي مبرمج آخر (أو لك مستقبلًا) أن هذه القيمة «مقدّسة» ولا تُعدَّل.

## 3.11 جدول مقارنة شامل: أنواع البيانات الأساسية

| النوع | الكلمة المفتاحية | مثال | هل قابل للتعديل (Mutable)؟ |
|---|---|---|---|
| عدد صحيح | `int` | `25` | ❌ لا (Immutable) |
| عدد عشري | `float` | `3.14` | ❌ لا (Immutable) |
| نص | `str` | `"مرحبا"` | ❌ لا (Immutable) |
| منطقي | `bool` | `True` | ❌ لا (Immutable) |
| فارغ | `NoneType` | `None` | ❌ لا (Immutable) |

> 💡 **ملاحظة تمهيدية مهمة**: لاحظ أن كل الأنواع التي تعلمناها اليوم هي **Immutable** (غير قابلة للتعديل المباشر). هذا المفهوم سيصبح شديد الأهمية في الفصل الثامن عندما نتعرّف على القوائم (Lists) التي هي **Mutable** (قابلة للتعديل) - الفرق بينهما سيُفسّر سلوكيات غريبة ستواجهها لاحقًا إن لم تفهم هذا الأساس جيدًا من الآن.
