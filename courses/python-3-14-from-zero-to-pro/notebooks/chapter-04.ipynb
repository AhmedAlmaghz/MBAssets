## 4.3 الحروف الخاصة (Escape Characters)

أحيانًا تحتاج لتضمين رموز خاصة داخل النص يصعب كتابتها مباشرة. Python توفر «رموز هروب» لهذا:

| الرمز | المعنى | مثال |
|---|---|---|
| `\n` | سطر جديد (New Line) | `"سطر1\nسطر2"` |
| `\t` | مسافة تبويب (Tab) | `"اسم:\tمحمد"` |
| `\\` | شرطة مائلة فعلية | `"C:\\Users"` |
| `\"` | علامة اقتباس مزدوجة داخل نص مزدوج | `"قال: \"مرحبا\""` |
| `\'` | علامة اقتباس مفردة داخل نص مفرد | `'لا تقل: \'لا\''` |

````python
message: str = "السطر الأول\nالسطر الثاني"
print(message)
# الناتج:
# السطر الأول
# السطر الثاني
````

### 🔧 النصوص الخام (Raw Strings)

أحيانًا تريد أن تُعامَل الشرطة المائلة `\` **حرفيًا** دون أي تفسير خاص - خصوصًا عند التعامل مع مسارات الملفات على Windows أو التعبيرات النمطية (Regex) لاحقًا:

````python
# بدون raw string - مشكلة!
path: str = "C:\new_folder"  # \n هنا تُفسَّر كسطر جديد! خطأ غير متوقع
print(path)
# الناتج: C:
# ew_folder  ← خطأ!

# مع raw string - الحل
path: str = r"C:\new_folder"
print(path)
# الناتج: C:\new_folder  ← صحيح تمامًا
````

> ⚠️ **تحذير عملي مهم**: هذا الخطأ شائع جدًا بين المبتدئين الذين يعملون على Windows ويتعاملون مع مسارات الملفات. دائمًا استخدم `r"..."` عند كتابة مسارات Windows، أو الأفضل - استخدم `pathlib` التي سنتعلمها في فصل لاحق وتتجنب هذه المشكلة بالكامل.

## 4.4 f-strings: الطريقة الحديثة والوحيدة التي يجب أن تتعلمها

### 🕰️ خلفية تاريخية سريعة (للفهم لا للحفظ)

عبر تاريخ Python، وُجدت 3 طرق لتنسيق النصوص ودمج المتغيرات فيها:

````python
name = "سارة"
age = 28

# الطريقة 1: % formatting (قديمة جدًا، نادرًا ما تُستخدم اليوم)
print("اسمي %s وعمري %d" % (name, age))

# الطريقة 2: .format() (كانت شائعة، أصبحت أقل استخدامًا)
print("اسمي {} وعمري {}".format(name, age))

# الطريقة 3: f-strings (الحديثة - من Python 3.6+)
print(f"اسمي {name} وعمري {age}")
````

> 🎯 **قرار تعليمي واضح**: في هذا الكتاب، سنستخدم **f-strings حصريًا تقريبًا**، لأنها الأسلوب الموصى به عالميًا اليوم - أوضح، أسرع تنفيذيًا، وأقل عرضة للأخطاء. ستجد الطريقتين القديمتين في كود آخرين أو مشاريع قديمة، لذا من المهم أن تتعرّف عليهما، لكن لا تستخدمهما في كودك الخاص.

### ✍️ الصيغة الأساسية لـ f-strings

ضع حرف `f` قبل علامة الاقتباس، وضع أي متغير داخل أقواس `{}`:

````python
name: str = "محمد"
age: int = 25

greeting: str = f"مرحبًا {name}, عمرك {age} سنة"
print(greeting)  # مرحبًا محمد, عمرك 25 سنة
````

### 🚀 قوة f-strings: التعبيرات الكاملة داخل الأقواس

الميزة الحقيقية لـ f-strings أنها لا تقبل فقط أسماء متغيرات، بل **أي تعبير Python صالح**:

````python
price: float = 49.99
quantity: int = 3

# حسابات مباشرة
print(f"الإجمالي: {price * quantity}")  # الإجمالي: 149.97

# استدعاء دوال
name: str = "محمد"
print(f"الاسم بحروف كبيرة: {name.upper()}")  # الاسم بحروف كبيرة: محمد (ملاحظة: upper لا تؤثر على العربية)

# شروط مباشرة (Ternary expressions)
age: int = 15
print(f"الحالة: {'قاصر' if age < 18 else 'بالغ'}")  # الحالة: قاصر
````

### 🎨 تنسيق متقدم: التحكم في عرض الأرقام

f-strings توفر صياغة قوية للتحكم في كيفية عرض الأرقام - مهارة ستحتاجها كثيرًا في أي تطبيق حقيقي:

````python
price: float = 1234.5678

# تحديد عدد الخانات العشرية
print(f"{price:.2f}")  # 1234.57 (تقريب لخانتين عشريتين)

# إضافة فاصلة آلاف
print(f"{price:,.2f}")  # 1,234.57

# عرض كنسبة مئوية
ratio: float = 0.856
print(f"{ratio:.1%}")  # 85.6%

# محاذاة النص (مفيد جدًا للجداول في الطرفية)
name: str = "محمد"
print(f"[{name:>10}]")   # [     محمد] (محاذاة يمين، عرض 10)
print(f"[{name:<10}]")   # [محمد     ] (محاذاة يسار، عرض 10)
print(f"[{name:^10}]")   # [   محمد   ] (محاذاة وسط، عرض 10)
````

### 🐛 ميزة تشخيصية رائعة (Python 3.8+): `=` داخل f-string

````python
x: int = 10
y: int = 20

print(f"{x=}, {y=}, {x + y=}")
# الناتج: x=10, y=20, x + y=30
````

> 💡 **نصيحة عملية قوية**: هذه الميزة ممتازة أثناء تصحيح الأخطاء (Debugging) - بدلاً من كتابة `print(f"x is {x}")` يدويًا، فقط اكتب `print(f"{x=}")` وستحصل على اسم المتغير وقيمته معًا تلقائيًا. استخدمها بكثرة أثناء تعلّمك لفهم ما يحدث في كودك.

## 4.5 الفهرسة (Indexing): الوصول لحرف معيّن

### 🔢 فهم النظام: يبدأ العد من صفر دائمًا!

في Python (وأغلب لغات البرمجة)، كل نص هو **سلسلة من الحروف**، وكل حرف له «فهرس» (رقم موضعه) يبدأ من **0** وليس من 1.

````python
word: str = "مرحبا"
#            م  ر  ح  ب  ا
# الفهرس:    0  1  2  3  4
````

````python
word: str = "مرحبا"

print(word[0])  # م  (الحرف الأول)
print(word[1])  # ر
print(word[4])  # ا  (الحرف الأخير - الفهرس 4، لأن الطول 5)
````

> ⚠️ **تحذير حرج - خطأ شائع جدًا**: البدء من 0 وليس 1 يُربك المبتدئين باستمرار. تذكّر دائمًا: **الفهرس الأول هو 0**، والفهرس الأخير هو **طول النص - 1**.

### 🔄 الفهرسة السالبة: ميزة ذكية في Python

Python تسمح بالعد من **النهاية** باستخدام أرقام سالبة:

````python
word: str = "مرحبا"

print(word[-1])  # ا  (آخر حرف)
print(word[-2])  # ب  (قبل الأخير)
````

> 💡 **لماذا هذا مفيد؟** تخيّل أنك تريد آخر حرف في نص طويل لا تعرف طوله بالضبط. بدل حساب الطول يدويًا، فقط استخدم `[-1]` مباشرة - أنيق وبسيط.

### ⚠️ خطأ شائع: تجاوز حدود النص

```python
word: str = "مرحبا"
print(word[10])  # ❌ IndexError: string index out of range
```

⚠️ **للقراءة فقط** — خطأ مقصود للتوضيح، لذا لا يُنفَّذ.

**الحل**: دائمًا تأكد أن الفهرس الذي تطلبه موجود فعليًا ضمن طول النص (سنتعلم `len()` بعد قليل).

## 4.6 التقطيع (Slicing): استخراج جزء من النص

التقطيع يسمح لك باستخراج **جزء** من النص، وليس حرفًا واحدًا فقط:

### ✍️ الصيغة الأساسية

```python
text[start:end:step]
```

🧾 **للقراءة فقط** — مثال مقتطف يعتمد على `end`، `start`، `step` من مثال سابق.

- `start`: الفهرس الذي يبدأ منه (مُضمَّن)
- `end`: الفهرس الذي ينتهي عنده (**غير** مُضمَّن - نقطة مهمة جدًا!)
- `step`: القفزة بين كل عنصر (اختياري، الافتراضي 1)

### 🎯 أمثلة متدرجة

````python
text: str = "البرمجة ممتعة"

print(text[0:8])    # البرمجة  (من الفهرس 0 حتى 7، الفهرس 8 غير مُضمَّن)
print(text[9:])     # ممتعة    (من الفهرس 9 حتى النهاية)
print(text[:8])     # البرمجة  (من البداية حتى الفهرس 7)
print(text[:])      # البرمجة ممتعة  (نسخة كاملة من النص)
````

> ⚠️ **تحذير مهم جدًا - أكثر الأخطاء شيوعًا في التقطيع**: الفهرس `end` **غير مُضمَّن** في النتيجة. هذا يُربك المبتدئين باستمرار:

````python
text: str = "Python"
print(text[0:3])  # Pyt  (ليس Pyth!) - لأن الفهرس 3 غير مُضمَّن
````

### 🔄 استخدام step للقفز بين الحروف

````python
text: str = "0123456789"

print(text[::2])    # 02468  (كل حرف ثانٍ)
print(text[1::2])   # 13579  (كل حرف ثانٍ، بدءًا من الفهرس 1)
print(text[::-1])   # 9876543210  (عكس النص بالكامل!)
````

> 💡 **نصيحة عملية مشهورة**: `text[::-1]` هي الطريقة الأكثر شيوعًا وأناقة في Python لعكس أي نص بالكامل - احفظ هذا النمط جيدًا، ستستخدمه كثيرًا.

## 4.7 النصوص غير قابلة للتعديل (Immutable) - مفهوم حاسم

### 🔒 الفهم الجوهري

تذكر من الفصل السابق: النصوص (`str`) هي نوع بيانات **Immutable** - أي **لا يمكن تعديلها مباشرة** بعد إنشائها.

```python
word: str = "مرحبا"
word[0] = "M"  # ❌ TypeError: 'str' object does not support item assignment
```

⚠️ **للقراءة فقط** — خطأ مقصود للتوضيح، لذا لا يُنفَّذ.

### 🤔 لكن ألسنا «نُعدّل» النصوص باستمرار؟

نعم، لكن ما يحدث فعليًا هو إنشاء نص **جديد بالكامل**، وليس تعديل النص الأصلي في مكانه:

````python
word: str = "مرحبا"
new_word: str = word.upper()  # لا يُعدّل word، بل يُنشئ نصًا جديدًا تمامًا

print(word)      # مرحبا  (لم يتغيّر!)
print(new_word)  # (نسخة معدّلة جديدة)
````

> 💡 **لماذا هذا مهم فعليًا؟** هذا يُفسّر لماذا دائمًا تحتاج لإعادة إسناد النتيجة إلى متغير (نفس المتغير أو جديد) عند استخدام methods تُغيّر النص - الـ method لا تُغيّر الأصل أبدًا:

````python
name: str = "محمد"
name.upper()       # ⚠️ لا يفعل شيئًا فعليًا! النتيجة تُهدَر
print(name)        # محمد  (لم يتغيّر!)

name = name.upper()  # ✅ الطريقة الصحيحة - إعادة الإسناد
print(name)           # محمد (بأحرف كبيرة إن كانت إنجليزية)
````

## 4.8 أهم Methods النصوص (الأكثر استخدامًا عمليًا)

Python توفر عشرات الـ methods للنصوص، لكن لن نُغرقك بكلها. إليك **الأكثر استخدامًا فعليًا** في المشاريع الحقيقية:

### 🔠 تغيير حالة الأحرف

````python
text: str = "Hello World"

print(text.upper())       # HELLO WORLD
print(text.lower())       # hello world
print(text.title())       # Hello World (كل كلمة تبدأ بحرف كبير)
print(text.capitalize())  # Hello world (أول حرف فقط كبير)
````

### ✂️ إزالة المسافات الزائدة

````python
text: str = "   مرحبا بالعالم   "

print(text.strip())   # "مرحبا بالعالم" (إزالة المسافات من الطرفين)
print(text.lstrip())  # "مرحبا بالعالم   " (من اليسار فقط)
print(text.rstrip())  # "   مرحبا بالعالم" (من اليمين فقط)
````

> 💡 **استخدام عملي شديد الأهمية**: `.strip()` أساسية جدًا عند معالجة مدخلات المستخدم من `input()`، لأن المستخدمين كثيرًا ما يكتبون مسافات زائدة عن غير قصد:

```python
name: str = input("اسمك: ").strip()
```

🧾 **للقراءة فقط** — يحتاج إدخالًا يدويًا.

### 🔍 البحث داخل النص

````python
text: str = "البرمجة بلغة Python ممتعة"

print("Python" in text)        # True
print("Java" in text)          # False
print(text.find("Python"))     # 11 (الفهرس حيث بدأت الكلمة)
print(text.find("Java"))       # -1 (غير موجودة)
print(text.count("ة"))         # عدد مرات تكرار الحرف
````

> 🔄 **مقارنة مهمة: `in` مقابل `.find()`**

| الطريقة | متى تستخدمها |
|---|---|
| `in` | عندما تحتاج فقط معرفة: موجود أم لا؟ (يُعيد bool) |
| `.find()` | عندما تحتاج معرفة **أين** بالضبط (يُعيد رقم الفهرس، أو -1) |

### 🔄 الاستبدال

````python
text: str = "أنا أحب Java"
new_text: str = text.replace("Java", "Python")
print(new_text)  # أنا أحب Python
````

### ✂️ التقسيم والدمج (من أهم الـ methods على الإطلاق)

````python
sentence: str = "التفاح,الموز,البرتقال"

# تقسيم النص إلى قائمة (سنتعلم القوائم بعمق في الفصل الثامن)
fruits: list[str] = sentence.split(",")
print(fruits)  # ['التفاح', 'الموز', 'البرتقال']

# العملية العكسية: دمج قائمة في نص واحد
joined: str = " - ".join(fruits)
print(joined)  # التفاح - الموز - البرتقال
````

> 💡 **تطبيق عملي حقيقي جدًا**: `.split()` و `.join()` هما من أكثر الـ methods استخدامًا في معالجة البيانات الحقيقية (مثل قراءة ملفات CSV التي سنتعلمها لاحقًا).

### ✅ التحقق من محتوى النص

````python
print("12345".isdigit())      # True - كلها أرقام؟
print("Hello".isalpha())      # True - كلها حروف؟
print("Hello123".isalnum())   # True - حروف أو أرقام فقط؟
print("   ".isspace())        # True - كلها مسافات؟
print("Hello".startswith("He"))  # True
print("Hello".endswith("lo"))    # True
````

## 4.11 جدول مرجعي شامل: أهم Methods النصوص

| Method | الوظيفة | مثال |
|---|---|---|
| `.upper()` | تحويل لأحرف كبيرة | `"abc".upper()` → `"ABC"` |
| `.lower()` | تحويل لأحرف صغيرة | `"ABC".lower()` → `"abc"` |
| `.strip()` | إزالة المسافات من الطرفين | `" hi ".strip()` → `"hi"` |
| `.split(sep)` | تقسيم لقائمة | `"a,b".split(",")` → `["a","b"]` |
| `.join(list)` | دمج قائمة لنص | `"-".join(["a","b"])` → `"a-b"` |
| `.replace(old,new)` | استبدال جزء | `"hi".replace("h","H")` → `"Hi"` |
| `.find(sub)` | إيجاد موضع | `"hello".find("l")` → `2` |
| `.count(sub)` | عدّ التكرارات | `"aaa".count("a")` → `3` |
| `.startswith(s)` | يبدأ بـ؟ | `"hi".startswith("h")` → `True` |
| `.endswith(s)` | ينتهي بـ؟ | `"hi".endswith("i")` → `True` |
| `.isdigit()` | كلها أرقام؟ | `"123".isdigit()` → `True` |
| `.isalpha()` | كلها حروف؟ | `"abc".isalpha()` → `True` |
| `len(text)` | طول النص | `len("hello")` → `5` |

## 4.14 مشروع تطبيقي: «محلل النصوص الذكي»

### 📋 المهمة

أنشئ ملفًا باسم `text_analyzer.py` يقوم بتحليل أي نص يُدخله المستخدم، ويعرض تقريرًا كاملاً عنه:

**المتطلبات:**
1. طلب نص من المستخدم (جملة أو فقرة)
2. حساب: عدد الحروف (بدون مسافات)، عدد الكلمات، عدد المسافات
3. إيجاد أطول كلمة في النص
4. تحويل النص لصيغة «عنوان» (كل كلمة تبدأ بحرف كبير - إن كان إنجليزيًا)
5. عرض النص معكوسًا بالكامل
6. طباعة تقرير منسّق بشكل احترافي وجذاب

### 💻 حل نموذجي كامل

<details>
<summary>✅ اضغط لرؤية الحل الكامل (حاول بنفسك أولاً!)</summary>

```python
# text_analyzer.py
"""برنامج تحليل النصوص - يعرض إحصائيات شاملة عن أي نص مُدخَل."""

# استقبال النص من المستخدم
text: str = input("أدخل نصًا لتحليله: ").strip()

# تقسيم النص لكلمات لحساب الإحصائيات
words: list[str] = text.split()

# الحسابات الأساسية
total_chars: int = len(text.replace(" ", ""))
word_count: int = len(words)
space_count: int = text.count(" ")

# إيجاد أطول كلمة
longest_word: str = max(words, key=len) if words else ""

# تحويلات النص
title_case: str = text.title()
reversed_text: str = text[::-1]

# طباعة التقرير
print("\n" + "=" * 40)
print("📊  تقرير تحليل النص")
print("=" * 40)
print(f"📝 النص الأصلي: {text}")
print(f"🔤 عدد الحروف (بدون مسافات): {total_chars}")
print(f"📦 عدد الكلمات: {word_count}")
print(f"␣  عدد المسافات: {space_count}")
print(f"🏆 أطول كلمة: {longest_word}")
print(f"🔠 بصيغة العنوان: {title_case}")
print(f"🔄 النص معكوسًا: {reversed_text}")
print("=" * 40)
```

🧾 **للقراءة فقط** — يحتاج إدخالًا يدويًا.

**ملاحظة على الحل**: استخدمنا `max(words, key=len)` لإيجاد أطول كلمة - وهذا مثال تشويقي بسيط على دالة `max()` مع معامل `key`، وهو مفهوم سنتعمّق فيه لاحقًا في فصل الدوال المتقدمة. لا تقلق إن بدا غامضًا الآن.
